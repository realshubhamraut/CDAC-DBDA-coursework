````python
import nbformat as nbf

def md_to_ipynb(md_path, ipynb_path):
    with open(md_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    nb = nbf.v4.new_notebook()
    cells = []
    inside_code_block = False
    code_lines = []
    markdown_lines = []

    for line in lines:
        if line.strip().startswith("```python"):
            inside_code_block = True
            if markdown_lines:
                cells.append(nbf.v4.new_markdown_cell(''.join(markdown_lines)))
                markdown_lines = []
        elif line.strip().startswith("```") and inside_code_block:
            inside_code_block = False
            cells.append(nbf.v4.new_code_cell(''.join(code_lines)))
            code_lines = []
        elif inside_code_block:
            code_lines.append(line)
        else:
            markdown_lines.append(line)

    # add any leftover markdown
    if markdown_lines:
        cells.append(nbf.v4.new_markdown_cell(''.join(markdown_lines)))

    nb['cells'] = cells

    with open(ipynb_path, 'w', encoding='utf-8') as f:
        nbf.write(nb, f)

    print(f"✅ Notebook created successfully: {ipynb_path}")

# Usage
md_to_ipynb("python_basics.md", "python_basics.ipynb")
````
