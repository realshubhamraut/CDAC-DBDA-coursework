```python
# 55. Merge two DataFrames
df55a = pd.DataFrame({'key': ['A', 'B', 'C'], 'value1': [1, 2, 3]})
df55b = pd.DataFrame({'key': ['A', 'B', 'D'], 'value2': [4, 5, 6]})
print(df55a)
print(df55b)
pd.merge(df55a, df55b, on='key', how='inner')
```

```python
# 56. Left join
df56a = pd.DataFrame({'key': ['A', 'B', 'C'], 'value1': [1, 2, 3]})
df56b = pd.DataFrame({'key': ['A', 'B', 'D'], 'value2': [4, 5, 6]})
pd.merge(df56a, df56b, on='key', how='left')
```

```python
# 57. Right join
df57a = pd.DataFrame({'key': ['A', 'B', 'C'], 'value1': [1, 2, 3]})
df57b = pd.DataFrame({'key': ['A', 'B', 'D'], 'value2': [4, 5, 6]})
pd.merge(df57a, df57b, on='key', how='right')
```

```python
# 58. Outer join
df58a = pd.DataFrame({'key': ['A', 'B', 'C'], 'value1': [1, 2, 3]})
df58b = pd.DataFrame({'key': ['A', 'B', 'D'], 'value2': [4, 5, 6]})
pd.merge(df58a, df58b, on='key', how='outer')
```
