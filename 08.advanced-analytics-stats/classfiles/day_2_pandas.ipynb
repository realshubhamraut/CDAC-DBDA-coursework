#### **Pandas Cheat Sheet — Most Used Functions**

####  Basic inspection
| Function | Purpose |
|---------|---------|
| df.head() | Show first rows |
| df.tail() | Show last rows |
| df.info() | Summary: datatypes + null counts |
| df.describe() | Numerical summary (mean, std, etc.) |
| df.shape | (rows, columns) |
| df.columns | List of column names |
| df.dtypes | Datatype of each column |
| df.index | Show index range |

---

####  Handling missing values
| Function | Purpose |
|---------|---------|
| df.isnull() | True/False for missing |
| df.isnull().sum() | Count missing values |
| df.dropna() | Remove rows with missing values |
| df.fillna(value) | Fill missing values |

---

####  Selecting & filtering
| Function | Purpose |
|---------|---------|
| df['col'] | Select column |
| df[['col1','col2']] | Select multiple columns |
| df.loc[row_label, col_label] | Select by label |
| df.iloc[row_index, col_index] | Select by index |
| df[df.col > 10] | Conditional filtering |
| df.query("col > 10") | SQL-like filtering |

---

####  Sorting
| Function | Purpose |
|---------|---------|
| df.sort_values('col') | Sort by column |
| df.sort_values(['col1', 'col2']) | Sort by multiple columns |

---

####  Handling duplicates
| Function | Purpose |
|---------|---------|
| df.duplicated() | Identify duplicates |
| df.drop_duplicates() | Remove duplicates |

---

####  Renaming & modifying columns
| Function | Purpose |
|---------|---------|
| df.rename(columns={'old':'new'}) | Rename columns |
| df.astype({'col':'int'}) | Change datatype |
| df.drop(columns=['col']) | Drop a column |

---

####  Combining data
| Function | Purpose |
|---------|---------|
| pd.concat([df1, df2]) | Vertical/Horizontal stacking |
| df.merge(df2, on='col') | SQL-style join |
| df.join(df2) | Join by index |

---

####  GroupBy & Aggregation
| Function | Purpose |
|---------|---------|
| df.groupby('col').mean() | Group & compute mean |
| df.groupby('col').agg(['mean','sum','count']) | Multi aggregation |
| df.groupby(['A','B']).size() | Count rows per group |

---

####  Value counts & unique
| Function | Purpose |
|---------|---------|
| df.col.value_counts() | Count of each category |
| df.col.nunique() | Number of unique values |
| df.col.unique() | List of unique values |

---

####  Pivoting
| Function | Purpose |
|---------|---------|
| df.pivot(index='A', columns='B', values='C') | Reshape table |
| df.pivot_table(index='A', columns='B', values='C', aggfunc='sum') | Pivot with aggregation |

---

####  Statistics (EDA)
| Function | Purpose |
|---------|---------|
| df.col.mean() | Mean |
| df.col.median() | Median |
| df.col.mode() | Mode |
| df.col.std() | Standard deviation |
| df.col.var() | Variance |
| df.col.min(), df.col.max() | Range |
| df.corr() | Correlation matrix |
| df.cov() | Covariance matrix |

---
