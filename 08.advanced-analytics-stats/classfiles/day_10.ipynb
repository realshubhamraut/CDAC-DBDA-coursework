```python
"""
TWO SAMPLE PROPORTION TEST

- this test is used to compare the difference of two sample proportions
against a claim value and establish whether the difference between them is 
statistically significant or not

"""

# I claim that CDAC Mumbai has atleast 12% more engineers as compared to pune

# Ho: prop(M) - prop(P) >= 0.12
# Ha: prop(M) - prop(P) < 0.12

CL = 0.95
LoS = 0.05

# In a nbatch of 70 students in Mumbai , 45 are engineers
# In a nbatch of 100 students in Mumbai , 60 are engineers

engg = np.array([45,60])
counts = np.array([70,100])

ssp.proportions_ztest(engg,counts,value=0.12,alternative='smaller')
# => (np.float64(-1.018633307307723), np.float64(0.15418854277526844))
# since p-value is greater than LoS hence we do not reject the Ho

(45/70) - (60/100)
# => 0.04285714285714293
```

```python
# chi square test

# single categorical variables with 
# the counts obtained for different categories or levels received from the sample need to be compared against the claimed pattern, weather the diffeerence is statistically significant or not
```

```python
pattern = np.array([0.35, 0.3,0.15,0.12,0.08])
obs_counts = np.array([120, 105, 80,50,30])
```

```python
sum(obs_counts)
```

```python
exp_counts = pattern * sum(obs_counts)
```

```python
exp_counts
```

```python
scipy.stats.chisquare(obs_counts, f_exp= exp_counts)
```

```python
# above we can conclude that the counts do not meet the given pattern
```

```python
ts_array = (obs_counts - exp_counts) ** 2/ (exp_counts)
```

```python
ts_array  # so here the hyundai is the most statistically significant with 8.572
```

```python
sum(ts_array)
```
