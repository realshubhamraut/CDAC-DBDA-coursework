```python
df = pd.read_excel('08.advanced-analytics-stats/data/CDAC_DataBook.xlsx', sheet_name = 'Health')
```

```python
df.head()
```

```python
df2 = df.melt(id_vars=['Country', 'Disease'], var_name = 'Year',  value_name = 'Cases')

# id_vars are the columns that we want to keep intact 
# var_name = columns we are melting
# value_name = organizing of the melted columns valuesinto the new column (here 'cases')


# so here values are reorganized by changing the order 
```

```python
df3 = df2.pivot_table(columns='Disease', values='Cases',index=['Country', 'Year'])

# pivot_table is for widening / boradening the table, its exactly opposite for melt

# so basically we are putting the every unique value that comes under the disease to the new individual column

# remember the values in cases comes from the disease column

```
