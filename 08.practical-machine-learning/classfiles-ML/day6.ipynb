```python
df = pd.read_excel('data/SalesData.xlsx')

reg_dummy = pd.get_dummies(df.Region, drop_first=True).astype(int)
df.drop('Region', axis=1, inplace=True)
df = pd.concat([df,reg_dummy], axis=1)

```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(df.drop('Sales',axis=1), df.Sales, random_state=10, test_size=0.1)
```

```python
import numpy as np
```

```python
import pandas as pd
```

```python
#inverse relationship between learning rate and n_estimators
```

```python
xgb_model = XGBRegressor(n_estimators=1000, learning_rate=0.01)
xgb_model.fit(x_train, y_train)
y_pred = xgb_model.predict(x_test)
rmse = np.sqrt(np.mean((y_test - y_pred)**2))
print("RMSE: %f" % (rmse))
#RMSE: 1033.211456

```

```python
gbr = GradientBoostingRegressor()
gbr.fit(x_train,y_train)
print(gbr.score(x_train, y_train))
print(gbr.score(x_test, y_test))
```

```python
my_params={
    'loss':['squared_error', 'absolute_error','huber'],
    'learning_rate': [0.001,0.005,0.01],#,0.03,0.05,0.10,0.2],
    'n_estimators' : [10,30],#,50,100],
    #'criterion': ['friedman_mse', 'squared_error'],
    'max_depth':[2,3,4,5],
    #'min_samples_split' : [2,3,4,5],
    #'min_samples_leaf' : [2,3,4,5]
}

from sklearn.model_selection import GridSearchCV

grid=GridSearchCV(gbr,param_grid=my_params,cv=5,scoring='accuracy')
grid.fit(x_train,y_train)
grid.best_params_
```

```python
xgb = XGBRegressor()
mod1 = xgb.fit(x_train, y_train)
print(xgb.score(x_train, y_train))
print(xgb.score(x_test, y_test))

xgb = XGBRegressor(learning_rate=0.01, reg_lambda=1, reg_alpha=0.5,gamma=2,max_depth=3)
mod1 = xgb.fit(x_train, y_train)
print(xgb.score(x_train, y_train))
print(xgb.score(x_test, y_test))
```

```python
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
model = gnb.fit(x_train, y_train)
print(gnb.score(x_train, y_train))
print(gnb.score(x_test, y_test)) 
```

```python
from sklearn.ensemble import AdaBoostRegressor
abr = AdaBoostRegressor(n_estimators=20, learning_rate=0.05, loss='linear')
mod1 = abr.fit(x_train, y_train)
print(abr.score(x_train, y_train))
print(abr.score(x_test, y_test))
```
