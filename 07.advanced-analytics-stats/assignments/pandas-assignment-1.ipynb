```python
import pandas as pd
import numpy as np
```

```python
# 1. Create and display a one-dimensional array-like object
s = pd.Series([1, 2, 3, 4, 5, 6])
s
```

```python
# 2. Convert Series to Python list
list(s)
```

```python
# 3. Add, subtract, multiply and divide two Series
s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 9])
print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```

```python
# 4. Compare elements of two Series
s3 = pd.Series([2, 4, 6, 8, 10])
s4 = pd.Series([1, 3, 5, 7, 10])
print(s3 == s4)
print(s3 > s4)
print(s3 < s4)
```

```python
# 5. Convert dictionary to Series
d = {'a': 100, 'b': 200, 'c': 300, 'd': 400, 'e': 800}
pd.Series(d)
```

```python
# 6. Convert NumPy array to Series
arr = np.array([10, 20, 30, 40, 50])
pd.Series(arr)
```

```python
# 7. Change data type of Series
s7 = pd.Series([100, 200, 'python', 300.12, 400])
print(s7)
pd.to_numeric(s7, errors='coerce')
```

```python
# 8. Convert first column of DataFrame to Series
df = pd.DataFrame({'col1': [1, 2, 3, 4, 7, 11], 'col2': [4, 5, 6, 9, 5, 0], 'col3': [7, 5, 8, 12, 1, 11]})
print(df)
df['col1']
```

```python
# 13. Create subset based on condition
s13 = pd.Series(range(11))
print(s13)
s13[s13 <= 5]
```

```python
# 14. Change order of index
s14 = pd.Series([1, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'])
print(s14)
s14.reindex(['B', 'A', 'C', 'D', 'E'])
```

```python
# 15. Mean and standard deviation
s15 = pd.Series([1, 2, 3, 4, 5, 6, 7, 8, 9, 5, 3])
print(s15)
print(s15.mean())
print(s15.std())
```

```python
# 16. Items not present in another series
sr1 = pd.Series([1, 2, 3, 4, 5])
sr2 = pd.Series([2, 4, 6, 8, 10])
print(sr1)
print(sr2)
sr1[~sr1.isin(sr2)]
```

```python
# 17. Items not common in both series
pd.concat([sr1[~sr1.isin(sr2)], sr2[~sr2.isin(sr1)]]).reset_index(drop=True)
```

```python
# 18. Compute min, 25th, median, 75th, max
s18 = pd.Series(np.random.randn(20))
print(s18)
np.percentile(s18, [0, 25, 50, 75, 100])
```

```python
# 19. Frequency counts
s19 = pd.Series(np.random.randint(0, 10, 40))
print(s19)
s19.value_counts()
```

```python
# 20. Replace with most frequent value
s20 = pd.Series([3, 1, 1, 3, 2, 2, 1, 2, 3, 1, 2, 2, 2, 3, 3])
print(s20)
print(s20.value_counts().head(2))
most_freq = s20.value_counts().index[0]
s20.apply(lambda x: x if x == most_freq else 'Other')
```

```python
# 21. Positions of multiples of 5
s21 = pd.Series([1, 9, 8, 6, 9, 7, 1, 1, 1])
print(s21)
np.where(s21 % 5 == 0)[0]
```

```python
# 22. Extract items at given positions
s22 = pd.Series(np.random.randint(0, 10, 22))
print(s22)
s22.iloc[[0, 2, 6, 11, 21]]
```

```python
# 23. Get positions of items in another series
s23a = pd.Series(range(1, 11))
s23b = pd.Series([1, 3, 5, 7, 10])
print(s23a)
print(s23b)
[pd.Index(s23a).get_loc(i) for i in s23b]
```

```python
# 24. First and last character to uppercase
s24 = pd.Series(['php', 'python', 'java', 'c#'])
print(s24)
s24.apply(lambda x: x[0].upper() + x[1:-1] + x[-1].upper() if len(x) > 1 else x.upper())
```

```python
# 25. Number of characters in each word
s25 = pd.Series(['Php', 'Python', 'Java', 'C#'])
print(s25)
s25.apply(len)
```

```python
# 26. Difference of differences
s26 = pd.Series([1, 3, 5, 8, 10, 11, 15])
print(s26)
print(s26.diff().tolist())
print(s26.diff().diff().tolist())
```

```python
# 27. Convert date strings to timeseries
s27 = pd.Series(['01 Jan 2015', '10-02-2016', '20180307', '2014/05/06', '2016-04-12', '2019-04-06T11:20'])
print(s27)
pd.to_datetime(s27)
```

```python
# 28. Get day of month, day of year, week number, day of week
dates = pd.to_datetime(s27)
print([d.day for d in dates])
print([d.dayofyear for d in dates])
print([d.week for d in dates])
print([d.day_name() for d in dates])
```

```python
# 29. Convert year-month to dates with specific day
s29 = pd.Series(['Jan 2015', 'Feb 2016', 'Mar 2017', 'Apr 2018', 'May 2019'])
print(s29)
pd.to_datetime(s29 + ' 11')
```

```python
# 30. Filter words with at least two vowels
s30 = pd.Series(['Red', 'Green', 'Orange', 'Pink', 'Yellow', 'White'])
print(s30)
s30[s30.apply(lambda x: sum([c in 'aeiouAEIOU' for c in x]) >= 2)]
```

```python
# 31. Euclidean distance between two series
s31a = pd.Series(range(1, 11))
s31b = pd.Series([11, 8, 7, 5, 6, 5, 3, 4, 7, 1])
print(s31a)
print(s31b)
np.linalg.norm(s31a - s31b)
```

```python
# 32. Positions neighboured by smaller values
s32 = pd.Series([1, 8, 7, 5, 6, 5, 3, 4, 7, 1])
print(s32)
np.where((s32 > s32.shift(1)) & (s32 > s32.shift(-1)))[0]
```

```python
# 33. Replace white spaces with least frequent character
s33 = 'abc def abcdef icd'
print(s33)
freq = pd.Series(list(s33.replace(' ', ''))).value_counts()
print(freq)
s33.replace(' ', freq.index[-1])
```

```python
# 34. Compute autocorrelations
s34 = pd.Series(np.random.randn(15))
print(s34)
[s34.autocorr(lag=i).round(2) for i in range(1, 11)]
```

```python
# 35. Display all Sundays of given year
pd.date_range('2020-01-01', '2020-12-31', freq='W-SUN')
```

```python
# 36. Convert series to dataframe with index as column
s36 = pd.Series(range(5), index=list('ABCDE'))
s36.to_frame().reset_index()
```

```python
# 37. Stack two series vertically and horizontally
s37a = pd.Series(range(10))
s37b = pd.Series(list('pqrstuvwxy'))
print(s37a)
print(s37b)
pd.DataFrame({0: s37a, 1: s37b})
```

```python
# 38. Check equality of two series
s38a = pd.Series([1, 8, 7, 5, 6, 5, 3, 4, 7, 1])
s38b = pd.Series([1, 8, 7, 5, 6, 5, 3, 4, 7, 1])
print(s38a)
print(s38b)
s38a == s38b
```

```python
# 39. Index of first occurrence of smallest and largest value
s39 = pd.Series([1, 3, 7, 12, 88, 23, 3, 1, 9, 0])
print(s39)
print(s39.idxmin())
print(s39.idxmax())
```

```python
# 40. Check inequality of dataframe and series
df40 = pd.DataFrame({'W': [68.0, 75.0, 86.0, 80.0, np.nan], 'X': [78.0, 75.0, np.nan, 80.0, 86.0], 
                     'Y': [84, 94, 89, 86, 86], 'Z': [86, 97, 96, 72, 83]})
s40 = pd.Series([68.0, 75.0, 86.0, 80.0, np.nan])
print(df40)
print(s40)
df40.ne(s40, axis=0)
```
