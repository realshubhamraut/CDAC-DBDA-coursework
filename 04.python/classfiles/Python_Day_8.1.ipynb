```python
#Doubly linked List:
#operations:Insert left, insert right, delete left, delete right, insert after, delete specific, print
class Linked_list:
    def __init__(self):
        self.root=None
    def insert_left(self,n):
        if self.root==None:
            self.root=n;
        else:
            n.right=self.root#1
            self.root.left=n#2
            self.root=n;#3
    def insert_right(self,n):
        if self.root==None:
            self.root=n;
        else:
            t=self.root#1
            while t.right!=None:#2
                t=t.right
            t.right=n#3
            n.left=t#4

    def delete_left(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root#1
            self.root=self.root.right#2
            self.root.left=None#3
    def delete_right(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            t=self.root#1
            while t.right!=None:#2
                t=t.right
            if t==self.root:#single node
                self.root=None
            else:
                t2=t.left#3
                t2.right=None
            print(f"{t.data} deleted")
    def print_elements(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n!=None:
                print(f"|{n.data}|",end='->')
                n=n.right
    def reverse_print_elements(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n.right!=None:
                n=n.right
            #n is on right most
            while n!=None:
                print(f"|{n.data}|",end='->')
                n=n.left
```

```python
class Node:
	def __init__(self,data):
		self.data=data
		self.right=self.left=None
```

```python
class Tree:
    #Object of tree is created, root should be created as a member.
    def __init__(self):
        self.root=None
    #Should return reference of a root from the class to the script.
    def get_root(self):
        return self.root

    def insert(self,r,n):#Reference to the root and a new node.
        if self.root==None:
            self.root=n#The first node becomes the root.
        else:
            if n.data<r.data:
                if r.left==None:
                    r.left=n
                    print("Inserted")
                else:
                    self.insert(r.left,n)
            else:#right side
                if r.right==None:
                    r.right=n
                    print("Inserted")
                else:
                    self.insert(r.right,n)
    def inorder(self,r):
        if r!=None:
            self.inorder(r.left)#L
            print(r.data,end="-")#P
            self.inorder(r.right)#R

    def count_nodes(self,r):
        if r==None:
            return 0
        return 1+self.count_nodes(r.left)+self.count_nodes(r.right)

    def sum_nodes(self,r):
        if r==None:
            return 0
        return r.data+self.sum_nodes(r.left)+self.sum_nodes(r.right)

    def count_leaf_nodes(self,r):
        if r==None:
            return 0
        if r.left==None and r.right==None:
            return 1
        return self.count_leaf_nodes(r.left)+self.count_leaf_nodes(r.right)
    def searching_key(self,r,key):#true if found else false
        if self.root==None:
            print("Empty Tree")
            return False
        else:
            if r==None:
                return False
            else:
                if key==r.data:
                    return True
                elif key<r.data:
                    return self.searching_key(r.left,key)
                else:
                    return self.searching_key(r.right,key)


```

```python
import collections
import sys

# Define INF as a symbolic constant, though the Java code uses 999
INF = 999

class GraphDemo:
    """
    Python class equivalent to the Java GraphDemo structure.
    Implements graph creation via adjacency matrix, printing, and DFS/BFS traversals.

    Note: DFS and BFS methods are strictly translated to check for edge weight == 1,
    as done in the original Java code (if(g[source][i]==1)).
    """

    def __init__(self):
        # Instance variables equivalent to Java's v, g, visited
        self.v = 0              # Number of nodes
        self.g = []             # Adjacency matrix (list of lists)
        self.visited = []       # Visited array/list (0: unvisited, 1: visited)

    def createGraph(self, nodes):
        """Initializes and interactively populates the adjacency matrix."""
        self.v = nodes
        self.g = [[0 for _ in range(self.v)] for _ in range(self.v)]
        self.visited = [0] * self.v

        print("\n--- Interactive Graph Creation ---")
        print(f"Graph size: {self.v} vertices (V0 to V{self.v-1}).")

        for i in range(self.v):
            for j in range(self.v):
                print(f"Enter value for v{i} to v{j} ({INF} for infinity):")
                try:
                    # Input logic is strictly translated
                    self.g[i][j] = int(input(":"))
                except ValueError:
                    print("Invalid input. Setting edge weight to 0.")
                    self.g[i][j] = 0

    def printG(self):
        """Prints the adjacency matrix in a Java-style formatted output."""
        print("\n--- Adjacency Matrix ---")
        for i in range(self.v):
            row_output = ""
            for j in range(self.v):
                # Use tab (\t) for Java-like spacing
                row_output += str(self.g[i][j]) + "\t"
            print(row_output)
        print("-" * 30)

    def resetvisited(self):
        """Resets the visited array for a new traversal."""
        for i in range(self.v):
            self.visited[i] = 0 # 0 means unvisited

    def DFS(self, source):
        """
        Recursive Depth-First Search.
        Note: Assumes a connection if g[source][i] == 1.
        """
        self.visited[source] = 1
        print("V" + str(source))

        for i in range(self.v):
            # Check for neighbor AND unvisited, strictly using the Java logic g[source][i] == 1
            if self.g[source][i] == 1 and self.visited[i] != 1:
                self.DFS(i)

    def BFS(self, source):
        """
        Iterative Breadth-First Search using a deque (queue).
        Note: Assumes a connection if g[element][i] == 1.
        """
        queue = collections.deque() # Python's efficient queue

        self.visited[source] = 1
        queue.append(source) # enqueue

        print("\nBFS Path:")

        while queue: # not empty
            element = queue.popleft() # dequeue
            print("V" + str(element), end="-")

            for i in range(self.v):
                # Check for neighbor AND unvisited, strictly using the Java logic g[element][i] == 1
                if self.g[element][i] == 1 and self.visited[i] != 1:
                    self.visited[i] = 1 # visited
                    queue.append(i) # enqueue
        print()


if __name__ == "__main__":

    # Get graph size
    try:
        size_input = input("Enter size of graph (number of nodes, e.g., 5):").strip()
        graph_size = int(size_input) if size_input else 5
        if graph_size <= 0: graph_size = 5
    except ValueError:
        print("Invalid input for size. Defaulting to 5.")
        graph_size = 5

    # Create and populate the graph
    obj = GraphDemo()
    obj.createGraph(graph_size)
    obj.printG()

    # --- Run DFS ---
    # Assuming start node 0 as in Java main method
    start_node = 0
    if obj.v > 0 and 0 <= start_node < obj.v:
        obj.resetvisited()
        print("\n--- Traversal Results ---")
        print("DFS:")
        obj.DFS(start_node)

        # --- Run BFS ---
        obj.resetvisited()
        print("\nBFS:")
        obj.BFS(start_node)
    elif obj.v > 0:
        print(f"\nTraversal skipped. Please choose a start node between 0 and {obj.v-1}.")
```
