```python

def bubblesort(a):
    for i in range(len(a)-1):
        for j in range(len(a)-1):
            if a[j]>a[j+1]:
                a[j],a[j+1]=a[j+1],a[j]
```

```python
l=[3,1,8,9,2,4,6,5]
print(l)
bubblesort(l)
```

```python
def selection_sort(a):
    for i in range(len(a)-1):
        min= a[i]
        pos=i
        for j in range(i+1, len(a)):
            if a[j] < min:
                min = a[j]
                pos=j
        a[pos], a[i] =a[i],min
```

```python
def insertion_sort(a):
    for i in range(len(a)-1):
        key = a[i+1]
        j = i+1
        while j > 0 and a[j-1] > key:
            a[j] = a[j-1]
            j -= 1
        a[j] = key
```

```python
def quick_sort(a,start, end):
    i, j, pivot = start, end, start

    while i < j:
        while a[i] < a[pivot]:
            i += 1
        while a[j] > a[pivot]:
            j -= 1

        if i < j:
            a[i], a[j] = a[j], a[i]

    if i < end:
        quick_sort(a,i + 1, end)
    if start < j:
        quick_sort(a,start, j - 1)
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    return merge(left_half, right_half)


def merge(left, right):
    merged = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1

    while i < len(left):
        merged.append(left[i])
        i += 1

    while j < len(right):
        merged.append(right[j])
        j += 1

    return merged
```

```python
def binarysearch(key,a,start,end):
    if start<=end:
        mid=(start+end)//2
        if a[mid]==key:
            print ("element found at:",mid)
        elif key<a[mid]:
            return binarysearch(key,a,start,mid-1)#left half
        elif key>a[mid]:
            return binarysearch(key,a,mid,end)#left half
    else:
        print("not found")
```
