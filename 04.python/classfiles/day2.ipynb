```python
# check if given num is armstrong

num = int(input("enter the number: "))
temp_num = num
armstrong_sum = 0
while temp_num > 0:
    digit = temp_num % 10
    armstrong_sum = armstrong_sum + digit ** 3
    temp_num = temp_num // 10

if num == armstrong_sum:
    print(f"{num} is armstrong")
else:
    print(f"{num} is not armstrong")
```

```python
# Write a program to check if a given positive integer is a Perfect Number.
# A perfect number is a positive integer that is equal to the sum of its proper positive divisors (that is, the sum of its positive divisors excluding the number itself).

# Example 1:
# Input: 6
# Divisors (excluding 6): 1,2,3
# Sum: 1+2+3=6
# Output: True

# Example 2:
# Input: 28
# Divisors (excluding 28): 1,2,4,7,14
# Sum: 1+2+4+7+14=28
# Output: True

n = int(input("Enter a positive integer: "))
div_sum = 0
for i in range(1, n):
    if n % i == 0:
        div_sum += i

if div_sum == n:
    print(f"{n} is a perfect number")
else:
    print(f"{n} is not a perfect number")
```

```python
# Use above code to implement a billing system where the user would enter:
# - Number of teas
# - Number of coffees
# - Number of waters
# - Number of bun muskas
# Calculate total bill amount to be paid and bill should only display items which are ordered. Also, tea is 10 rupees, coffee is 40 rupees, water is 20 rupees, and bun maska is 100.

# Prices
prices = {
    "Tea": 10,
    "Coffee": 40,
    "Water": 20,
    "Bun Maska": 100
}

# Take order quantities
qty_tea = int(input("Enter number of teas: "))
qty_coffee = int(input("Enter number of coffees: "))
qty_water = int(input("Enter number of waters: "))
qty_bunmaska = int(input("Enter number of bun maskas: "))

# Prepare bill
bill_items = []
total = 0

if qty_tea > 0:
    amt = qty_tea * prices["Tea"]
    bill_items.append(f"Tea x {qty_tea} = {amt}")
    total += amt
if qty_coffee > 0:
    amt = qty_coffee * prices["Coffee"]
    bill_items.append(f"Coffee x {qty_coffee} = {amt}")
    total += amt
if qty_water > 0:
    amt = qty_water * prices["Water"]
    bill_items.append(f"Water x {qty_water} = {amt}")
    total += amt
if qty_bunmaska > 0:
    amt = qty_bunmaska * prices["Bun Maska"]
    bill_items.append(f"Bun Maska x {qty_bunmaska} = {amt}")
    total += amt

print("\n--- Bill ---")
for item in bill_items:
    print(item)
print(f"Total: {total} rupees")
```
