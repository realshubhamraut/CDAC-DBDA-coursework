```python
#passing by value:changes Made local and seen locally.
#all basic data members (int,float,str,bool,complex) are passed by value
def swap1(no1,no2):
    print(f"\tstart in function no1:{no1} no2:{no2}")
    no1,no2=no2,no1
    print(f"\tend in function no1:{no1} no2:{no2}")
#passing by reference:#changes Made local are seen globally.
def doubler(numbers):
    print(f"\tstart in function numbers are:{numbers}")
    for index in range(len(numbers)):
        numbers[index]*=2
    print(f"\tend in function numbers are:{numbers}")
#all basic data structures (list,tuple,dict,set) are passed by ref
```

```python
no1,no2=100,200
print(f"start in main script no1:{no1} no2:{no2}")
swap1(no1,no2)
print(f"end in main script no1:{no1} no2:{no2}")
```

```python
#Generate Fibonacci series up to 50 elements using generator concept in Python.
#0,1,1,2,3,5,8....  50th term
def fibo(n):
    f0,f1=0,1
    for i in range(n+1):
        if i<=1:
            yield i
        else:
            fn=f0+f1
            f0,f1=f1,fn
            yield fn
```

```python
value=fibo(50)
```

```python
print(next(value))
```
