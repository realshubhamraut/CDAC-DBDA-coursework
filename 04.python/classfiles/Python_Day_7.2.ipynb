```python
class DS_Example:
    def __init__(self):
        self.stack=[]
        #Created an empty stack.
    def push(self,e):
        self.stack.append(e)
        #print(f"{e} pushed on stack")
    def pop(self):
        return (self.stack.pop())#default
        #remove and return
    def peek(self):
        return self.stack[-1]
        #only return
    def print_stack(self):#in LIFO
        for index in range(len(self.stack)-1,-1,-1) :
            print(self.stack[index])
    def is_empty(self):
        return len(self.stack)==0
```

```python
#Convert decimal to binary:
#input:decimal number o/p:binary sequence(string)
#convert_bin(10):--->1010
def convert_bin(no):
    bin=[]
    while no>0:
        digit=str(no%2)
        bin.append(digit)
        no=no//2
    return "".join(reversed(bin))
```

```python
Interview Coding Question: Maximum Gold Profit in a Fixed Window
 (Sliding Window)

Scenario:

You are a data analyst optimizing a gold trading strategy. You have a list of daily gold prices for a period of several weeks. A crucial constraint of your strategy is that once you buy a certain quantity of gold, you must sell that same quantity within a consecutive time window of exactly 7 days (i.e., if you buy on day i, you must sell on a day j such that j−i<7).

Challenge:

Write a Python function max_profit_in_fixed_window(prices, window_size) that determines the single transaction (one buy, one sell) that yields the maximum profit while strictly adhering to the fixed window constraint.

Input:

prices: A list of positive integers or floats representing the daily price of gold.

window_size: An integer representing the maximum allowed duration of the trade (e.g., 7 days).

Output:
The function should return a tuple containing:

The maximum achievable profit (max_profit).

The 1-based index of the optimal buy day (buy_day_index).

The 1-based index of the optimal sell day (sell_day_index).

Example:

If prices = [10, 8, 15, 12, 20] and window_size = 3 (meaning a maximum index difference of 2):

Buy Day 2 (Price 8), Sell Day 3 (Price 15). Profit: 7. (Index diff: 1, valid)

Buy Day 3 (Price 15), Sell Day 5 (Price 20). Profit: 5. (Index diff: 2, valid)

Buy Day 2 (Price 8), Sell Day 5 (Price 20). Profit: 12. (Index diff: 3, invalid since 3≮3)

The expected output would be: (7, 2, 3) (representing a profit of 7, buying on Day 2, and selling on Day 3).
```

```python

def best_deal(data):
    profit=0
    buy,sell=0,0
    for i in range(len(data)):
        for j in range(i+1,len(data)):
            if data[j]-data[i]> profit:
                buy=i
                sell=j
                profit=data[sell]-data[buy]
    return profit,buy,sell
```

```python
profit,day_buy,day_sell=best_deal([10, 8, 15, 12, 20])
print(f"profit of {profit} got by buying on {day_sell} and selling no {day_buy} ")
```

```python
#Priorty queue:
#Priority queue can be ascending or descending.
#In ascending priority queue on DQ, we get data from smallest to largest.
#In descending priority queue, we get data value from largest to smallest.
#example [10,2,45,3,13,20]
#For ascending priority queue, data:2,3,10,13,20,45
#Descending priority queue, data:45,20,13,10,3,2
#Implementation purpose: You will be implementing Enqueue,Dequeue, print_queue, and is_empty.

class DS_Example:
    def __init__(self):
        self.queue=[]
        #Created an empty queue.
    def enqueue(self,e):#entry on queue
        self.queue.append(e)
        self.queue.sort()
        #print(f"{e} pushed on stack")
    def dequeue(self):
        return (self.queue.pop(0))#delete from start :0
        #remove and return
    def print_queue(self):#in LIFO
        for item in self.queue:
            print(item,sep=" - ");
    def is_empty(self):
        return len(self.queue)==0
```

```python
#Linked List
#Collection of nodes interconnected in a sequential manner.
#Node: a structure which consists of data and reference part.
```

```python
class Node:
	def __init__(self,data):
		self.data=data
		self.next=None
```

```python
from logging import root
from logging import root
from logging import root
#linked List:
#operations:Insert left, insert right, delete left, delete right, insert after, delete specific, print
class Linked_list:
    def __init__(self):
        self.root=None
    def insert_left(self,n):
        if self.root==None:
            self.root=n;
        else:
            n.next=self.root
            self.root=n;
    def insert_right(self,n):
        if self.root==None:
            self.root=n;
        else:
            t=self.root#go to last node
            while t.next!=None:
                t=t.next
            t.next=n#connect .next of last to n

    def delete_left(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root#1
            self.root=self.root.next#2
    def delete_right(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            t=t2=self.root#1
            while t.next!=None:
                t2=t
                t=t.next
            t2.next=None#3
            print(f"{t.data} deleted")
    def print_elements(self):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n!=None:
                print(f"|{n.data}|",end='->')
                n=n.next
    def search_elements(self,key):
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n!=None and n.data!=key:
                n=n.next
            if n==None:
                print("Element not found")
            else:
                print("Element found")
    def insert_after(self,key,e):#after is ref to search ,e is element to insert
        if self.root==None:
            print("\nEmpty list")
        else:
            n=self.root
            while n!=None and n.data!=key:
                n=n.next
            if n==None:
                print("Element not found")
            else:
                print("Element found")
                new_node=Node(e)
                new_node.next=n.next
                n.next=new_node
                print("inserted....")
```

```python
o=Linked_list()
o.insert_left(Node(10))
o.insert_left(Node(5))
o.insert_left(Node(1))
print()
o.print_elements()
o.insert_right(Node(100))
o.insert_right(Node(50))
o.insert_right(Node(1000))
print()
o.print_elements()
```
