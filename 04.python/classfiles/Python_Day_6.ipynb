```python
class student:
    #static member/class member
    student_no=202500#Use class dot to access this.
    def __init__(self,gender="unknown",name="un-named"):
        print(f"Human object created {id(self)}")
        self.gender=gender
        self.name=name
        student.student_no+=1
        self.roll_no=student.student_no
        print(f"Student regitered with roll no:{self.roll_no}")
    def get_roll_no(self):
        return self.roll_no
    def update(self):
        print("Old record:")
        print(self)
        print("Enter new details:")
        name=input("Enter name:")
        if name!="":
            self.name=name
        gender=input("Enter gender:")
        if gender!="":
            self.gender=gender
        print("Updated record:")
        print(self)

    def __str__(self):
        return f"Name:{self.name}\tGender:{self.gender}\tRoll_Number:{self.roll_no}"
        #modify str to print Name:<name> Gender:<gender> roll_no:<roll_no>
```

```python
def menu():
    print("-------------------------")
    print("Student Management System")
    print("-------------------------")
    print("1.Register a student")
    print("2.Search a student")
    print("3.Update a student")
    print("4.Delete a student")
    print("5.Print student list")
    print("0.Exit")

#dict for storing n students
s={}#empty dict
#menu
while True:
    menu()
    choice=int(input(":"))
    if choice==1:
        print("Welcome to student registration")
        name=input("Enter name:")
        gender=input("Enter gender:")
        obj_s=student(gender,name)#constructor
        s[obj_s.get_roll_no()]=obj_s#store
        #   202501------------->s_obj
    elif choice==5:
        for student_rollno in s:
            print(s[student_rollno])
    elif choice==2:
        roll=int(input("Enter roll number:"))#ask for roll no,check if there in
        if roll in s:#dict if so all display
            print("Record found")
            print(s[roll])
        else:
            print(f"{roll} not found")#else not found to be shown
    elif choice==3:
        roll=int(input("Enter roll number:"))#ask for roll no,check if there in
        if roll in s:#dict if so all display
            print("Record found")
            s[roll].update()
            #s[202501]===>object of student 202501 rollno
        else:
            print(f"{roll} not found")#else not found to be shown
    elif choice==4:
        roll=int(input("Enter roll number:"))#ask for roll no,check if there in
        if roll in s:#dict if so all display
            print("Record found")
            del s[roll]
            print("Record deleted")
        else:
            print(f"{roll} not found")#else not found to be shown
    elif choice==0:
        print("Exiting...........")
        break
    else:
        print("In-Valid option selected.")
```

```python
class Human:
    def set_Human(self,name,gender):
        self.__name=name
        self.__gender=gender
    def disply_Human(self):
        print(f"Name:{self.__name}\nGender:{self.__gender}")
class Education(Human):
    def set_Education(self,degree):
        self.__degree=degree
    def disply_education(self):
        print(f"\nDegree is:{self.__degree}")
class Job(Education):
    def set_Job(self,name,gender,degree,salary):
        self.__salary=salary
        self.set_Education(degree)
        self.set_Human(name,gender)
    def disply_Job(self):
        self.disply_Human()
        self.disply_education()
        print(f"\nSalary is:{self.__salary}")

```

```python
#multiple inheritance:
#Not Supported by Java due to ambiguity, which means unclear whose property should be taken, specially when multiple superclasses have exact properties to share.
#Python handles first-come, first-taken. In multiple inheritance, the first method is taken, and other methods are rejected in case of ambiguity.
class A:
    def my(self):
        print("My from A")
class B:
    def my(self):
        print("My from B")
class C(B,A):
    def myC(self):
        print("My from C")
```

```python
obj=C()
obj.myC()
obj.my()
```

```python
#override: when a child class re-codes a parent's method, it will reject the parent method and use its own method called "overwrite".
class Parent:
    def language(self):
        print("Parent:Hindi")
    def eat(self):
        print("Parent: Bhaji Chapati")
#need of override: change as per need.
#Old method needs to be reconstructed as per need.
```

```python
class Child(Parent):
    def eat(self):#Overriding eat of parent by recoding it.
        print("Child: Pizza, burger, fast food, Chinese.")
```

```python
class BankAccount:
    def __init__(self,amount):
        self.__amount=amount

    def deposit(self,amount):
        if amount>0:
            self.__amount+=amount
        else:
            print(f"{amount} can not be -ve")
        print(f"New balance is:{self.__amount}")

    def withdraw(self,amount):
        if self.__amount-amount<0:
            print("Insufficient Balance")
        else:

            if self.__amount-amount<3000:
                self.__amount-=500
                print("-500 charged for minimum balance.")
                print(f"New balance is:{self.__amount}")
            self.__amount-=amount
            print(f"New balance is:{self.__amount}")

```

```python
class Student:
    def __init__(self,name,*grades):
        self.__name=name
        self.__grades=grades
    #display name and grades
    def __str__(self):
        return f"Name:{self.__name} and grades are {self.__grades}"
    def cal_average(self):
        return sum(self.__grades)/len(self.__grades)
    def find_highest_grade(self):
        return max(self.__grades)
    def find_lowest_grade(self):
        return min(self.__grades)


```

```python
s=Student("amar",90,80,50,80,88,66)
```

```python
print(s)
```

```python
s=Student("amar",90,80,50,80,88,66)
print(s)
print("Avrage is:",s.cal_average())
print("Min is:",s.find_lowest_grade())
print("Max is:",s.find_highest_grade())
```
