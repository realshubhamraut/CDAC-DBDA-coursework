Data Structure
It's a way of storing data such that depending on the application, it takes minimum time to update, read, and access.
python's built in DS
All data structures are heterogeneous(any/all types) in nature.
list,set,dict:dynamic(can grow or shrink at runtime)
tuple:static(fixed size)
list:[]
	hetrogeneous,dynamic,indexed access,can have duplicates
	Max used by programmer(array,sequence,list,collection)
tuple:()
	hetrogeneous,static,indexed access,can have duplicates
	Once created, cannot be changed.
	Max used by python for security(passing data)
set:{}
	hetrogeneous,dynamic,not indexed access(auto iteration
     supported),can not have duplicates.
	Max use for uniqueness/filter to unique value
dict:{key:value}
      hetrogeneous,dynamic,not indexed access but key accessed,can not have duplicates key but can have duplicate values.
	Max used for mapping value to data

```python
#Keep accepting elements and keep storing them in a list
#till the user gives the first blank.
#Produce the sum of all the numbers entered in the list.
data_list=[]
while True:#infinite while loop
    data=input("Enter:")
    if data=="":
        print("Blank given so stopping")
        break
    data_list.append(int(data))
print(f"Total data in list {len(data_list)} : {data_list}")
print(f"Sum is: {sum(data_list)}")
```

```python
#Keep accepting elements and keep storing them in a list
#till the user gives the first blank.
#Produce the sum of all the numbers entered in the list.
#First, print all elements lesser than the average.
#Then, print all the elements greater than the average.
data_list=[]
while True:#infinite while loop
    data=input("Enter:")
    if data=="":
        print("Blank given so stopping")
        break
    data_list.append(int(data))
print(f"Total data in list {len(data_list)} : {data_list}")
average=sum(data_list)/len(data_list)
print(f"Sum is: {sum(data_list)} and Average is:{average}")
print("\nElements Lesser than average")
for item in data_list:
    if item<average:
        print(item,end=",")
print("\nElements Greater than average")
for item in data_list:
    if item>=average:
        print(item,end=",")
```

```python
# Problem Definition
# Goal: Create an input text for a text summarization tool that is specifically designed to test its word count, unique word count, and individual word frequency/count features, with the final output of the word list being sorted alphabetically.

# Input Requirement: A sample paragraph where a significant number of words (especially common, short, or technical/thematic ones) are repeated multiple times to ensure the frequency counting mechanism is thoroughly tested. The paragraph should be cohesive enough to be analyzed as a single body of text.

# Desired Output: The text summarizer should analyze the input paragraph and provide:

# Total Word Count (e.g., 105 words).

# Total Unique Word Count (e.g., 35 unique words).

# A Word Frequency List showing each unique word and its count, sorted in alphabetical order.

# sample:
# The expert technical trainer, with the vast expertise in training, started the new career coaching training. The goal of the training was to train every fresher in the required skills. An expert trainer knows the value of a strong skill set for a promising career. This dedicated career training is the perfect platform to acquire that much-needed skill. The trainer, an expert in his career, said that the best skill a fresher can acquire is the skill to learn new skills. We train freshers for a better career; we train them to be experts.

# Expected Analysis Data (for your reference)
# Word	Count
# a	4
# acquire	2
# an	2
# and	1
# are	1
# be	1
# best	1
# better	1
# career	5
# can	1
# coaching	1
# dedicated	1
# every	1
# expert	5
# expertise	1
# for	2
# fresher	2
# freshers	2
# goal	1
# his	1
# in	2
# is	2
# learn	1
# much	1
# needed	1
# new	1
# of	2
# perfect	1
# platform	1
# promising	1
# required	1
# said	1
# set	1
# skill	6
# skills	2
# that	1
# the	10
# their	1
# them	1
# technical	1
# this	1
# to	5
# train	6
# trainer	3
# training	4
# value	1
# vast	1
# we	2
# with	1
```

```python
sample_lines='''
The expert technical trainer, with the vast expertise in training, started the new career coaching training. The goal of the training was to train every fresher in the required skills. An expert trainer knows the value of a strong skill set for a promising career. This dedicated career training is the perfect platform to acquire that much-needed skill. The trainer, an expert in his career, said that the best skill a fresher can acquire is the skill to learn new skills. We train freshers for a better career; we train them to be experts.
'''
sample_lines=sample_lines.replace(',','')
sample_lines=sample_lines.replace('.','')
sample_lines=sample_lines.replace(';','')
print(sample_lines)
sample_lines=sample_lines.lower()
word_list=sample_lines.split()
unique_word_list=sorted(set(word_list))#sorted by default returns list
print(f"Total word counts:{len(word_list)}")
print(f"Total unique word counts:{len(unique_word_list)}")
for word in unique_word_list:
    print(f"{word}\t\t{word_list.count(word)}")
#give longest word in para
word_length=0
longest_word=""
for word in unique_word_list:
    if len(word)>word_length:
        word_length,longest_word=len(word),word
print(f"\nLonget word is:{longest_word} length is {word_length}")
```
