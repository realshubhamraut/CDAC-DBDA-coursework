```python
Question 2: Employee Performance Rating

Problem:
An employee’s yearly performance is rated between 1 to 100. Based on the score, assign a grade:

90–100 → "Outstanding"

75–89 → "Excellent"

60–74 → "Good"

40–59 → "Average"

Below 40 → "Poor"
```

```python
# Employee Performance Rating

# Input employee score
score = int(input("Enter employee performance score (1-100): "))

# Check and assign grade
if 90 <= score:
    grade = "Outstanding"
elif 75 <= score:
    grade = "Excellent"
elif 60 <= score:
    grade = "Good"
elif 40 <= score:
    grade = "Average"
elif 0 <= score:
    grade = "Poor"
else:
    grade = "Invalid Score"  # handles values outside 0-100

# Output the grade
print("Employee Performance Rating:", grade)
```

```python
#continue
    When executed, it will skip the current pass and start the next pass.
#break
    When executed, it will stop the loop, but the program will continue.
```

```python
for i in range(1,11,1):
    if i==5:
        continue
    print(i)
print("code ends")
```

```python
for i in range(1,11,1):
    if i==5:
        break
    print(i)
print("code ends")
```
