```python
#Python does not support abstract classes. To make abstract classes possible, we need third-party packages.
#Abstract Class:is a class which has one or more abstract methods.
#Abstract method are methods without body
#Abstract method and class used for ? making implementation standard must to code needed methods:override is must
#steps:1: import from abc module ABC(class)and abstractmethod
from abc import ABC, abstractmethod
#step 2: inherit ABC as Parent class
class Animal(ABC):
    @abstractmethod
    def speak(self):#signature should match
        pass
    #step 3 create abstarct method with pass as body:no operation in method
```

```python
class Dog(Animal):
    def speak(self):
        return "bow bow"
class Crow(Animal):
    def speak(self):
        return "kaw kaw"
```

```python
d=Dog()
```

```python
d.speak()
```

```python
#An abstract method called `area` which will simply return `area`.
#Shape is inherited by circle and rectangle.
#Implement circle rectangle, create an object of circle rectangle and show overriding.
#Abstract class is compulsory.
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def set_radius(self,radius):
        self.radius=radius
    #def area(self):
     #   return 3.14*self.radius**2

```

```python
obj=Circle()
obj.set_radius(4.5)
obj.area()
```

```python
#Polymorphic Behaviour(Duck Typing)
#evey object behaves as per own method depending on who he/she/it is.
class Dog:
    def speak(self):
        return "bow bow"
class Crow:
    def speak(self):
        return "kaw kaw"
class Human:
    def speak(self):
        return "English only"
```

```python
def communication(e):
    print(e.speak())
```

```python
#depending on what/who speak would be called
d=Dog()
c=Crow()
h=Human()
```

```python
communication(h)
```
