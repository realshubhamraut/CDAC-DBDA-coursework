```python
#accept a number and reverse it
#i/p :123   o/p:321
#Extract one digit at a time use formula rno= rno*10+digit
number=int(input("Enter a number:"))#123
rno=0
while number>0:
    digit=number%10#123 3      12 2         1
    rno=rno*10+digit#0*10+3=3  3*10+2=32    32*10+1=321
    number=number//10 #12      1            0
print(f"Reverse number is {rno}")
```

```python
#check whether the given number is a palindrome or not. The input must be an integer
#i/p :121  o/p:121 yes it is palindrome
#1. Read number
#2. Copy to temp
#3. Destroy temp to Reverse the number
#4. Compare original and reverse
number=int(input("Enter a number:"))#123
t_number=number
rno=0
while t_number>0:
    digit=t_number%10#123 3      12 2         1
    rno=rno*10+digit#0*10+3=3  3*10+2=32    32*10+1=321
    t_number=t_number//10 #12      1            0
if number==rno:
    print(f"{number} is palindrome")
else:
    print(f"{number} is not palindrome")
```

```python
#Armstrong: 153---1^3+5^3+3^3=153
#Cube of individual digits added together is the number itself.
number=int(input("Enter a number:"))#153
t_number=number
sum=0
while t_number>0:
    digit=t_number%10#153 3 5 1
    sum=sum+digit**3      # 3^3+ 5^3 + 1^3
    t_number=t_number//10
if number==sum:
    print(f"{number} is armstrong")
else:
    print(f"{number} is not armstrong")
```

```python
#Print all Armstrong numbers from 100 to 1000.
for number in range(100,1000):#loop for 100 to 1000
    t_number=number
    sum=0
    while t_number>0:
        digit=t_number%10#153 3 5 1
        sum=sum+digit**3      # 3^3+ 5^3 + 1^3
        t_number=t_number//10
    if number==sum:
        print(f"{number} is armstrong")
```

```python
#Check whether the given number is prime or not.
#flag concept: flag=true---if something goes wrong false
number=int(input("Enter a number:"))#5
flag=True
for i in range(2,number):#2,3,4
    print(f"div by {i}")
    if number%i==0: #6%2==0
        flag=False  #Flase
        break       #stop
if flag:#if flag==True
    print(f"{number} is prime")
else:
    print(f"{number} is not prime")
```

```python
#Print all prime numbers from 1 to 100.
for number in range(2,100):
    flag=True
    for i in range(2,number):#2,3,4
        if number%i==0: #6%2==0
            flag=False  #Flase
            break       #stop
    if flag:#if flag==True
        print(f"{number} is prime")
```

```python
Write a program to check if a given positive integer is a Perfect Number. A perfect number is a positive integer that is equal to the sum of its proper positive divisors (that is, the sum of its positive divisors excluding the number itself).

Example 1:
Input: 6
Divisors (excluding 6): 1,2,3
Sum: 1+2+3=6
Output: True

Example 2:
Input: 28
Divisors (excluding 28): 1,2,4,7,14
Sum: 1+2+4+7+14=28
Output: True
```

```python
number=int(input("Enter a number:"))
sum=0
for i in range(1,number):
    if number%i==0:
        sum+=i
if sum==number:
    print(f"{number} is perfect number")
else:
    print(f"{number} is not perfect number")
```

```python
Use above code to implement a billing system where the user would enter:
- Number of teas
- Number of coffees
- Number of waters
- Number of bun maska
Calculate total bill amount to be paid and bill should only display items which are ordered. Also, tea is 10 rupees, coffee is 40 rupees, water is 20 rupees, and bun maska is 100.
```

```python
#menu driven
tea_cost,coffee_cost,water_cost,bm_cost=10.00,40.00,20.00,100.00
tea_qty=coffee_qty=water_qty=bm_qty=0
while True:
    #menu
    print("Apna Tea Stall")
    print("--------------")
    print("1.Tea")
    print("2.Coffee")
    print("3.Water")
    print("4.Bun Maska")
    print("0.Exit")
    print("--------------")
    choice=int(input("Enter:"))
    if choice==1:
        tea_qty+=int(input("tea selected enter quantity:"))
    elif choice==2:
        coffee_qty+=int(input("coffee selected enter quantity:"))
    elif choice==3:
        water_qty+=int(input("water selected enter quantity:"))
    elif choice==4:
        bm_qty+=int(input("bun maska selected enter quantity:"))
    elif choice==0:
        print("Thanks for visit")
        break#stops the loop
    else:#default
        print("Please check the option.")
#billing
print(f"          Bill           ")
print(f"-------------------------")
print(f"item   qty   cost   total")
if tea_qty>0:
    print(f"tea    {tea_qty}     {tea_cost}   {tea_qty*tea_cost}")
if coffee_qty>0:
    print(f"coffee {coffee_qty}     {coffee_cost}   {coffee_qty*coffee_cost}")
if water_qty>0:
    print(f"Water  {water_qty}     {water_cost}   {water_qty*water_cost}")
if bm_qty>0:
    print(f"tea    {bm_qty}     {bm_cost}   {bm_qty*bm_cost}")
print(f"-------------------------")
print(f"Total amount to pay{tea_qty*tea_cost+coffee_qty*coffee_cost+water_qty*water_cost+bm_qty*bm_cost}")

```

```python
line=input("Enter message:")
shift=int(input("Enter shifts:"))
encrypted_line=""
for char in line:
    if char>='a' and char<='z':
        encrypted_data=(ord(char)-ord('a')+shift)%26+ord('a')
              #       index       within 0 to 25  +97
        encrypted_character=chr(encrypted_data)
        encrypted_line+=encrypted_character
    else:
        encrypted_line+=char#spaces
print("Encrypted Message:",encrypted_line)
```

```python
line="aaaabbbaaaaabbbbbb"
compressed_line=""
i=j=0#i from o to last use while
while i<len(line)-1:
    alpha=line[i]#string at ith of line
    j=i+1#loop j from i+1 till not matching with ith
    while j<len(line) and line[i]==line[j]:
        j+=1
    compressed_line+=alpha+str(j-i)#read char from i and add difference (j-i)+1 as count
    i=j#shift i to jth position
print(compressed_line)



```
