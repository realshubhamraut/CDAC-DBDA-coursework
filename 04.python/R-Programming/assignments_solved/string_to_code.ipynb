```python
# 1. Count Vowels 
# Problem Definition: 
# Write a function to count the number of vowels (a, e, i, o, u) in a given string. The check should be case insensitive. 

s = input("enter thae string")
vowel_count ={'a':0, 'e':0, 'i':0, 'o':0, 'u':0}
for i in s.lower():
    if i in vowel_count:
        vowel_count[i] += 1

for vowel, count in vowel_count.items():
    print(f"{vowel}:{count}")
```

```python
# 3. Check Palindrome 
# Problem Definition: 
# Write a function to check if a string is a palindrome (reads the same forwards and backward).
# Ignore the case of the letters. 

s = input("enter string")
s_lower = s.lower()
rev_str = ""
i  = len(s_lower) - 1
while i>=0:
    rev_str = rev_str + s_lower[i]
    i -= 1

if s_lower == rev_str:
    print(f"{s} is palindrome")
else:
    print(f"{s} is not palindrome")

```

```python
# 9. Remove Duplicates (Preserve Order)
# Problem Definition:
# Remove all duplicate characters from a string, ensuring you keep the first occurrence of each character and
# maintain the original order.

text = "programming"
result = ""
for char in text:
    found = False
    for c in result:
        if c == char:
            found = True
            break
    if not found:
        result += char
print(result)
```

```python
# 11. Find First Substring Index
# Problem Definition:
# Find the starting index of the first occurrence of a smaller substring within a larger string. If the substring is not
# found, return 1. (You must implement the search logic; do not use the built-in 'find()' or "index)' methods).


text = "abracadabra"
substring = "cad"
result = -1
for i in range(len(text) - len(substring) + 1):
    match = True
    for j in range(len(substring)):
        if text[i + j] != substring[j]:
            match = False
            break
    if match:
        result = i
        break
print(result)
```

```python
# 17. String Compression
# Problem Definition:
# Implement a basic string compression method that replaces repeated characters with the character followed by its
# count. For example, "aaabbc" becomes "a3b2c1". Only return the compressed string if it is shorter than the original
# string; otherwise, return the original string.

text1 = "aaabbccccdde"
text2 = "abc"

def compress(text):
    if len(text) == 0:
        return text
    
    compressed = ""
    current_char = text[0]
    count = 1
    
    for i in range(1, len(text)):
        if text[i] == current_char:
            count += 1
        else:
            compressed += current_char + str(count)
            current_char = text[i]
            count = 1
    
    compressed += current_char + str(count)
    
    if len(compressed) < len(text):
        return compressed
    else:
        return text

print(compress(text1))
print(compress(text2))

```

```python
# 18. Check Prefix/Suffix
# Problem Definition:
# Determine if a string starts with a specific prefix AND ends with a specific suffix. You must implement the check
# logic; do not use the built-in startswith(' or 'endswith)" methods.

text = "SHUBHAM"
prefix = "SHU"
suffix = "AM"

# Check prefix
has_prefix = True
if len(prefix) <= len(text):
    for i in range(len(prefix)):
        if text[i] != prefix[i]:
            has_prefix = False
            break
else:
    has_prefix = False

# Check suffix
has_suffix = True
if len(suffix) <= len(text):
    start = len(text) - len(suffix)
    for i in range(len(suffix)):
        if text[start + i] != suffix[i]:
            has_suffix = False
            break
else:
    has_suffix = False

print(has_prefix and has_suffix)
```

```python
# 19. Interleave Strings
# Problem Definition:
# Combine two strings by alternating characters from each. Start with the first string. If one string is longer, append
# the remaining characters from the longer string to the result.

str1 = "abc"
str2 = "xyz123"
result = ""
i = 0
while i < len(str1) or i < len(str2):
    if i < len(str1):
        result += str1[i]
    if i < len(str2):
        result += str2[i]
    i += 1
print(result)
```

```python
# 20. Simple String Calculator
# Problem Definition:
# Evaluate a string that contains a very simple arithmetic exp (e.g., "10+5", "20-8") and return the integer
# result. Assume the exp only contains two positive integers and a single operator, which is either addition
# (+) or subtraction (-).


exp = "100-25"
opr_ind = -1
opr = ""
for i in range(len(exp)):
    if exp[i] == '+' or exp[i] == '-':
        opr_ind = i
        opr = exp[i]
        break

num1 = int(exp[:opr_ind])
num2 = int(exp[opr_ind + 1:])

if opr == '+':
    result = num1 + num2
else:
    result = num1 - num2

print(result)
```
