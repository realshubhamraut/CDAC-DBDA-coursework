```python
Check even or odd
What: Return "Even" or "Odd".


Approach: n % 2 == 0


Signature: def even_or_odd(n: int) -> str:


Sample input: n = 7


Sample output: "Odd"

```

```python

def even_or_odd(n: int) -> str:
    return "Even" if n%2==0 else  "odd"
even_or_odd(14)
```

```python
def max_of_three(no1, no2):
    return no1 if no1 > no2 else no2

max_of_three(12,15)
```
