```python
def reverse_str(s: str) -> str:
    s.split()
    s = s[::-1]
    return s

reverse_str('shubham raut')
```

```python
# Count vowels in a string

def count_vowels(s: str)-> int:
    count = 0
    s = s.lower()
    vowels = 'aeiou'
    splitd = s.split()
    for i in splitd:
        if s in vowels:
            count = count + 1
        return count
count_vowels('shubham')

```

```python
def line_reverse(line):
    words = line.strip().split()
    words = words[::-1]
    return ' '.join(words)

line_reverse('shubham is a boy')

```
