13) Remove duplicates from a list (preserve order)
What: Return list with duplicates removed but keep original order.


Approach: use seen set, append unseen items.


Signature: def remove_dups(lst: list) -> list:


Sample input: lst=[1,2,2,3,1]


Sample output: [1,2,3]


Edge cases: preserve order matters vs. set() which may reorder.


Complexity: O(n) time, O(n) space.


```python
def remove_dups(lst: list) -> list:
    new_list = []
    for i in lst:
        if i not in new_list:
            new_list.append(i)

    return new_list

remove_dups([12,12,12])
```

14) Second largest number in a list (distinct)
What: Return the second largest distinct value.


Approach: convert to set to dedupe then find max or sort.


Signature: def second_largest(lst: list):


Sample input: lst = [10,5,8,12,12]


Sample output: 10


Edge cases: fewer than 2 distinct numbers → return None or raise error.


Complexity: O(n) (or O(n log n) if sorting).

```python
def second_largest(lst: list):
    without_dups = []
    for i in lst:
        if i not in without_dups:
            without_dups.append(i)
    without_dups.sort()
    sec = without_dups[-2]

    return sec

second_largest([12,14,15])
```

19) Convert list of digits into concatenated number
What: [1,2,3] -> 123.


Approach: int("".join(map(str, lst))) (assumes list elements represent digits or numbers).


Signature: def list_to_number(digits: list) -> int:


Sample input: digits = [1,2,3]


Sample output: 123


Edge cases: elements >9 (e.g., [12,3]) — decide semantics (either join strings or treat as invalid).


Complexity: O(n).

```python
def list_to_num(li: list) -> int:
    store = ""
    for d in li:
        try:
            int_d = int(d)
        except Exception:
            raise TypeError("Enter the numbers only")
        
        if int_d < 0:
            raise ValueError("only positive nums are accepted")
        
        else:
            store += str(int_d)

    return int(store)

print(list_to_num([12,12,12]))
```

21) Factorial using recursion
What: Same factorial but recursive.


Approach: base case n==0 or n==1 -> 1, else n * fact(n-1).


Signature: def factorial_rec(n: int) -> int:


Sample input: n = 6


Sample output: 720


Edge cases: recursion depth for large n; n<0 raise error.


Complexity: O(n) time, O(n) recursion depth (space).

```python
def factorial_rec(n: int) -> int:
    if n < 0:
        raise ValueError("n must be a non-negative integer")
    if n == 0 or n == 1:
        return 1
    return n * factorial_rec(n - 1)

print(factorial_rec(6))  
```

22) Nth Fibonacci number (recursive or iterative)
What: Return nth Fibonacci. Define fib(0)=0, fib(1)=1.


Approach: Iterative or DP for performance (avoid naïve recursion).


Signature: def fibonacci(n: int) -> int:


Sample input: n = 6


Sample output: 8 (sequence: 0,1,1,2,3,5,8 — index 6 → 8)


Edge cases: negative n → error. Naïve recursion exponential; prefer iterative O(n) or matrix/exponentiation for O(log n).


Complexity: O(n) iterative, O(1) space (or O(log n) with fast exponentiation).

```python
def fibonacci(n: int) -> int:
    if n < 0:
        raise ValueError("n must be a non-negative integer")
    if n == 0:
        return 0
    if n == 1:
        return 1

    a, b = 0, 1 
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b


print(fibonacci(6))
```

23) Sum of digits (recursively)
What: Return sum of all digits of a non-negative integer.


Approach: base n==0 -> 0; else n%10 + sum_digits(n//10).


Signature: def sum_digits(n: int) -> int:


Sample input: n = 12345


Sample output: 15


Edge cases: negative numbers — take abs or reject. Recursion depth limited by digits ~ log10(n).


Complexity: O(d) where d is digit count.

```python
def sum_digits(n: int) -> int:
    if n < 0:
        raise ValueError("n must be a non-negative integer")
    if n == 0:
        return 0
    return n % 10 + sum_digits(n // 10)


print(sum_digits(12345))
```

24) Power function using recursion (x^y)
What: Compute x**y for non-negative integer y.


Approach: fast exponentiation (divide-and-conquer): if y even → (x^(y/2))^2, else x * x^(y-1).


Signature: def power(x, y: int):


Sample input: x=2, y=10


Sample output: 1024


Edge cases: y negative → return float 1/(x^|y|) or restrict to integers. x=0,y=0 undefined (decide behavior).


Complexity: O(log y) time with exponentiation by squaring.

```python
def power(x, y: int):
    if x == 0 and y == 0:
        raise ValueError("0^0 is undefined")
    
    if y == 0:
        return 1
    
    if y < 0:
        return 1 / power(x, -y)
        if y == 1:
        return x
    
    if y % 2 == 0:
        half = power(x, y // 2)
        return half * half
    else:

        return x * power(x, y - 1)


print(power(2, 10))  
```

25) GCD of two numbers (Euclidean algorithm, recursive)
What: Greatest common divisor.


Approach: gcd(a,b) = gcd(b, a % b) with gcd(a,0)=a.


Signature: def gcd(a: int, b: int) -> int:


Sample input: a=48, b=18


Sample output: 6


Edge cases: handle zeros, negative numbers use abs().


Complexity: O(log(min(a,b))) average.

```python
def gcd(a: int, b: int) -> int:
    a, b = abs(a), abs(b)

    if b == 0:
        return a
    return gcd(b, a % b)

print(gcd(48, 18)) 
```

26) Recursively print all elements of a list
What: Print/list elements using recursion (demonstrates recursion base-case).

Signature: def print_list_recursive(lst, i=0):


Sample input: lst=[10,20,30]


Sample output (printed lines):

 10
20
30


Edge cases: empty list → nothing printed. Recursion depth = len(list).


Complexity: O(n) time, O(n) space (recursion stack).

```python
def print_list_recursive(lst, i=0):
    if i >= len(lst):
        return
    print(lst[i])
    print_list_recursive(lst, i+1)


print_list_recursive([10, 20, 30])
```
