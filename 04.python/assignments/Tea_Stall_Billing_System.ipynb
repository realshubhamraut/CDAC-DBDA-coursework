```python

menu = {
    "cutting chai": 15.00,
    "masala chai": 25.00,
    "samosa": 20.00,
    "vada pav": 30.00,
    "biscuit packet": 10.00
}


membership_discount = {
    "gold": 0.10,
    "silver": 0.05,
    "none": 0.00
}


order_input = input("Enter your order (e.g., cutting chai 5 samosa 2): ").lower()
tokens = order_input.split()


order_dict = {}
i = 0
while i < len(tokens):

    if i + 1 < len(tokens) and tokens[i] + " " + tokens[i+1] in menu:
        item = tokens[i] + " " + tokens[i+1]
        qty = int(tokens[i+2])
        order_dict[item] = qty
        i += 3
    elif tokens[i] in menu:
        item = tokens[i]
        qty = int(tokens[i+1])
        order_dict[item] = qty
        i += 2
    else:
        i += 1

membership = input("Enter membership (GOLD/SILVER/NONE): ").lower()

subtotal = 0
item_details = []
for item, qty in order_dict.items():
    price = menu[item]
    total = price * qty
    subtotal += total
    item_details.append([item, qty, price, total])


discount_rate = membership_discount.get(membership, 0.00)
discount_amount = subtotal * discount_rate
taxable_total = subtotal - discount_amount


gst_rate = 0.05
gst_amount = taxable_total * gst_rate


final_amount = taxable_total + gst_amount


print("\n=============================================")
print("         TEA STALL BILLING SYSTEM")
print("=============================================")
print(f"{'ITEM':<20} {'QTY':<6} {'PRICE':<8} {'TOTAL':<8}")
print("---------------------------------------------")

for item, qty, price, total in item_details:
    print(f"{item.title():<20} {qty:<6} {price:<8.2f} {total:<8.2f}")

print("---------------------------------------------")
print(f"{'SUBTOTAL':<35} {subtotal:>8.2f}")
print(f"MEMBERSHIP: {membership.upper()} ({discount_rate*100:.2f}% Off)")
print(f"{'DISCOUNT APPLIED':<35} {-discount_amount:>8.2f}")
print("---------------------------------------------")
print(f"{'TAXABLE TOTAL':<35} {taxable_total:>8.2f}")
print(f"{'GST (5.00%)':<35} {gst_amount:>+8.2f}")
print("---------------------------------------------")
print(f"{'TOTAL AMOUNT DUE':<35} {final_amount:>8.2f}")
print("=============================================")
print("THANK YOU! VISIT AGAIN!")
```
