```python
#1.
class Car:
    def __init__(self, name, brand, year):
        self.name = name
        self.brand = brand
        self.year = year
    def display_info(self):
        print(f"name is {self.name}, year: {self.year}, brand{self.brand}")


c = Car("BMW", "XDW-1", 2002)
print(c.display_info())
```
