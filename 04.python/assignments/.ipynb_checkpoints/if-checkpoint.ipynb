```python
# 1. Write a program that takes a number as input and prints out whether it is an even or odd number.


n = int(input('enter a number: '))
if n%2 == 0:
    print('number is even')
else:
    print('number is odd')
```

```python
# 2. Write a program that takes a year as input and prints out whether it is a leap year or not.
year = int(input('enter year: '))

if ((year%4 ==0 and year %100!=0) or year % 400 == 0):
    print('leap year')
else:
    print('not a leap year')
```

```python
# 3. Write a program that takes a string as input and checks whether it is a palindrome or not.

st = input('enter string: ')
st = list(st)
if st[::-1] == st:
    print('its palindrome!')
else:
    print('not palindrome')
```

```python
# 4. Write a program that takes three numbers as input and prints out the largest one.

a,b,c = map(int, input("enter 3 numbers").split(','))

if a>b and a>c and a!=b!=c:
    print("a is largest")

elif b>a and b>c and a!=b!=c:
    print("b is greatest")

elif c>a and c>b and a!=b!=c:
    print("c is greatest")

elif a==b and a!=c:
    print('a and b is same')

elif a==c and a!=b:
    print('a and c is same')

elif b==c and b!=a:
    print('b and c is same')

else:
    print('a,b,c are equally same')
```

```python
# 5.Write a program that takes a sentence as input and counts the number of words in it.

string1 = input("enter a sentence")
string_li = string1.split()
len_string = len(string_li)

print(len_string)
```

```python
# 6. Write a program that takes a list of numbers as input and prints out the sum of all the numbers.


nums = list(map(int, input("Enter nums with space").split()))
print(nums)

num_sum = sum(nums)
print(num_sum)
```

```python
# 7.Write a program that takes a list of strings as input and prints out the longest string.

li_str = list(map(str, input("enter the string").split()))

if li_str:
    longest = li_str[0]
    for s in li_str:
        if len(s) > len(longest):
            longest = s
    print("The longest string is:", longest)
else:
    print("no string provided.")
```

```python
# 8. Write a program that takes two lists of numbers as input and prints out the common elements between the two lists.


li = list(map(int, input("Enter numbers").split()))

li2 = list(map(int, input("Enter list").split()))

for i in li:
    if i in li2:
        print(i)
```

```python
# 9. Write a program that takes a list of numbers as input and prints out the numbers in ascending order.


li = list(map(int, input("Enter the n umbers").split()))
li.sort()
print(li)
```
