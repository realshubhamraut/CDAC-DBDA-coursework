```python
#3. 
class BankAccount:
    def __init__(self,__balance):
        self.__balance = __balance

    def deposit(self, d_amt):
        self.__balance += d_amt
        print(f"the deposit money: {d_amt} and current balance {self.__balance}")
        
    def withdraw(self, w_amt):
        if w_amt <= self.__balance:
            self.__balance -= w_amt
            print(f"the withdraw money: {w_amt} and current balance {self.__balance}")
        else:
            print("insufficient funds")                    
bk = BankAccount(5000)
bk.deposit(500)
bk.withdraw(40000)
```

```python
#4

class Student:
    def __init__(self, name, *grades):
        self.name = name
        self.grades = grades

    def calculate_average(self):
        return sum(self.grades) / len(self.grades)

    def find_highest_grades(self):
        return max(self.grades)
    def find_lowest_grades(self):
        return min(self.grades)

s = Student("Shubham", 23,12,141)
print(f"average grades is: {s.calculate_average()}")
print(f"highest grade is: {s.find_highest_grades()}")
print(f"lowest grade is: {s.find_lowest_grades()}")
```
