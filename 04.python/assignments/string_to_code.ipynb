```python
# 1. Count Vowels 
# Problem Definition: 
# Write a function to count the number of vowels (a, e, i, o, u) in a given string. The check should be case insensitive. 

s = input("enter thae string")
vowel_count ={'a':0, 'e':0, 'i':0, 'o':0, 'u':0}
for i in s.lower():
    if i in vowel_count:
        vowel_count[i] += 1

for vowel, count in vowel_count.items():
    print(f"{vowel}:{count}")
```

```python
# 3. Check Palindrome 
# Problem Definition: 
# Write a function to check if a string is a palindrome (reads the same forwards and backward).
# Ignore the case of the letters. 


s = input("Enter a string: ")
s_lower = s.lower()  # make the check case-insensitive

# Reverse the string using a while loop
reversed_str = ""
i = len(s_lower) - 1
while i >= 0:
    reversed_str = reversed_str + s_lower[i]
    i -= 1

# Check if the original (lowercase) string is the same as the reversed version
if s_lower == reversed_str:
    print(f"'{s}' is a palindrome")
else:
    print(f"'{s}' is not a palindrome")

# ps - here we can't really iterate through every digits like we could on strings so

```
