```python
# 1. Car Class: Basic Information
# Create a `Car` class with attributes `make`,
# `model`, and `year`. Implement a method,
# `display_info()`, that prints all attributes in a
# readable format.



class Car:
    def __init__(self, make, model, year):
        self.make = make
        self,model = model
        self.year = year

    def display_info(self):
        print(f"the car manufacturer is {self.make}, the model is {self.model}, and the year is {self.year}")

c = Car("BMW", "AS-1")
c.display_info()


    
```
