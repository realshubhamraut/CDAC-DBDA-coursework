```python
# you are given an array of coin denominatoins and a target amount.
# Write a program to solve the compmlete coin change problem that includes finding the minimum coins needed counting all
# possible ways to make change and returning the actual coins combination used
# 1. min_coins(coins, amount) - find minimum number of coins needed to make the targetamount,
# if its not possible to make the amount, return -1 count_ways(coins, amount) - find the total amount of combinations not permutations
# 3. get_coin_combination(coin, amount) -returns one valid combo of coins
```

```python
def min_coins(coins, amount):

    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    
    for i in range(amount + 1):
        if dp[i] != float('inf'):
            for coin in coins:
                if i + coin <= amount:
                    dp[i + coin] = min(dp[i + coin], dp[i] + 1)
    
    return dp[amount] if dp[amount] != float('inf') else -1

def count_ways(coins, amount):

    dp = [0] * (amount + 1)
    dp[0] = 1
    
    for i in range(amount + 1):
        if dp[i] > 0:
            for coin in coins:
                if i + coin <= amount:
                    dp[i + coin] += dp[i]
    
    return dp[amount]


def get_coin_combination(coins, amount, current_combination = []):

    if amount == 0:
        return [current_combination[:]]
    combinations = []
    for coin in coins:
        if amount >= coin:
            new_combinations = get_coin_combination(coins, amount - coin, current_combination + [coin])
            combinations.extend(new_combinations)
    return combinations

coins = [3, 2, 5]
amount = 10
print(f"minimum coins: {min_coins(coins, amount)}")
print(f"total count ways: {count_ways(coins, amount)}")
print(f"total combinations: {get_coin_combination(coins, amount)}")

```
