```python
menu_items = {
    "cutting chai": 15,
    "masala chai": 25,
    "samosa": 20,
    "vada pav": 30,
    "biscuit packet": 10
}

# Get order input
order_input = input("Enter your order (e.g., cutting chai 5 samosa 2): ").lower()
user_order = order_input.split()

# Parse order into dictionary
order_dict = {}
i = 0
while i < len(user_order):
    # Check for multi-word items
    if i + 1 < len(user_order) and user_order[i] + " " + user_order[i+1] in menu_items:
        item_name = user_order[i] + " " + user_order[i+1]
        quantity = int(user_order[i+2])
        order_dict[item_name] = quantity
        i += 3
    elif user_order[i] in menu_items:
        item_name = user_order[i]
        quantity = int(user_order[i+1])
        order_dict[item_name] = quantity
        i += 2
    else:
        i += 1

# Calculate bill
print("\n--- YOUR ORDER ---")
total = 0
for item, qty in order_dict.items():
    if item in menu_items:
        price = menu_items[item]
        item_total = price * qty
        total += item_total
        print(f"{item.title()}: {qty} x ₹{price} = ₹{item_total}")
    else:
        print(f"{item.title()}: Not available")

print(f"\nTOTAL BILL: ₹{total}")
```

```python
def fibo(n):
    f0, f1 = 0, 1
    for i in range(n + 1):
        if i <= 1:
            yield i
        else:
            fn = f0 + f1
            f0, f1 = f1, fn
            yield f
```
