```python
# 📝 Problem Definition: User Contact File Generator
# 1. Goal
# The primary goal is to develop a program that securely collects Name, Email address, and Contact Number from the user, validates the input for basic correctness, and then permanently stores this data in a structured, readable format within a designated file.

# 2. Input
# The program must prompt the user to input the following three mandatory data fields:

# User Name: Full name (or preferred name).

# Email Address: A valid email address (e.g., user@example.com).

# Contact Number: A numerical string representing a phone number (e.g., 10 digits).

# 3. Processing and Constraints
# The program must perform the following actions and adhere to the specified constraints:

# Input Collection: Prompt the user clearly for each of the three data points.

# Basic Validation:

# Name: Must not be empty.

# Email: Must contain the '@' symbol and a '.' (dot) after the '@'.

# Number: Must only contain digits and be a reasonable length (e.g., 10 digits for standard mobile numbers).

# File Handling: The program must open a specified file (e.g., user_contacts.txt or a dynamically named file) in append mode to ensure existing data is not overwritten.

# Data Formatting: The collected and validated data must be formatted into a consistent, easily parsable structure before being written to the file. A suitable format could be a Comma-Separated Values (CSV) style or a label-based block format.

# 4. Output
# The program's output will consist of two main components:

# File Output (Persistent Storage): The collected data for each user is written to the output file in a clear, formatted manner.

# Example Formats:

# CSV: Amar Panchal,amar.career@example.com,9821601163

# Block:

# --- Contact Record ---
# Name: Amar Panchal
# Email: amar.career@example.com
# Number: 9821601163
# ----------------------
```

```python
def generate_contact():
    # Prompt for name until a non-empty string is entered
    while True:
        name = input("Your name: ").strip()
        if name:
            break
        else:
            print("Name cannot be empty. Please try again.")
            
    # Prompt for email until it passes basic validation
    while True:
        email = input("Enter email: ").strip()
        if "@" in email:
            parts = email.split("@")
            if len(parts) == 2 and "." in parts[1]:
                break
        print("Invalid email. Please ensure it contains '@' and a '.' after '@'. Try again.")
    
    # Prompt for mobile until a 10-digit numerical string is provided
    while True:
        contact = input("Enter mobile (10 digits): ").strip()
        if contact.isdigit() and len(contact) == 10:
            break
        else:
            print("Invalid mobile number. It must be exactly 10 digits. Please try again.")
    
    # Format record and append to file
    record = f"{name},{email},{contact}\n"
    with open("user_contacts.txt", "a") as file_object:
        file_object.write(record)
    
    print("Contact saved successfully.")

# Generate contact and then read/display the file content
generate_contact()

with open("user_contacts.txt", "r") as file_data:
    contacts = file_data.read()
    print(contacts)
```

```python
def get_non_empty_input(prompt):
    while True:
        value = input(prompt).strip()
        if value:
            return value
        print("Input cannot be empty. Please try again.")

def get_valid_email(prompt):
    while True:
        email = input(prompt).strip()
        if "@" in email:
            parts = email.split("@")
            if len(parts) == 2 and "." in parts[1]:
                return email
        print("Invalid email. Please enter a valid email address.")

def get_valid_contact(prompt):
    while True:
        contact = input(prompt).strip()
        if contact.isdigit() and len(contact) == 10:
            return contact
        print("Invalid contact number. It must be exactly 10 digits.")

def save_contact():
    name = get_non_empty_input("Enter your full name: ")
    email = get_valid_email("Enter your email address: ")
    contact = get_valid_contact("Enter your contact number (10 digits): ")

    record = (
        "\n--- Contact Record ---\n"
        f"Name: {name}\n"
        f"Email: {email}\n"
        f"Number: {contact}\n"
        "----------------------\n"
    )

    with open("user_contacts.txt", "a") as file:
        file.write(record)

    print("Contact saved successfully.\n")
    print("Current contacts in file:\n")
    with open("user_contacts.txt", "r") as file:
        print(file.read())

save_contact()
```

13) Remove duplicates from a list (preserve order)
What: Return list with duplicates removed but keep original order.


Approach: use seen set, append unseen items.


Signature: def remove_dups(lst: list) -> list:


Sample input: lst=[1,2,2,3,1]


Sample output: [1,2,3]


Edge cases: preserve order matters vs. set() which may reorder.


Complexity: O(n) time, O(n) space.


```python
def remove_dups(lst: list) -> list:
    new_list = []
    for i in lst:
        if i not in new_list:
            new_list.append(i)

    return new_list

remove_dups([12,12,12])
```

14) Second largest number in a list (distinct)
What: Return the second largest distinct value.


Approach: convert to set to dedupe then find max or sort.


Signature: def second_largest(lst: list):


Sample input: lst = [10,5,8,12,12]


Sample output: 10


Edge cases: fewer than 2 distinct numbers → return None or raise error.


Complexity: O(n) (or O(n log n) if sorting).

```python
def second_largest(lst: list):
    without_dups = []
    for i in lst:
        if i not in without_dups:
            without_dups.append(i)
    without_dups.sort()
    sec = without_dups[-2]

    return sec

second_largest([12,14,15])
```

19) Convert list of digits into concatenated number
What: [1,2,3] -> 123.


Approach: int("".join(map(str, lst))) (assumes list elements represent digits or numbers).


Signature: def list_to_number(digits: list) -> int:


Sample input: digits = [1,2,3]


Sample output: 123


Edge cases: elements >9 (e.g., [12,3]) — decide semantics (either join strings or treat as invalid).


Complexity: O(n).

```python
def list_to_num(li: list) -> int:
    store = ""
    for d in li:
        try:
            int_d = int(d)
        except Exception:
            raise TypeError("Enter the numbers only")
        
        if int_d < 0:
            raise ValueError("only positive nums are accepted")
        
        else:
            store += str(int_d)

    return int(store)

print(list_to_num([12,12,12]))
```

```python
import requests

def get_emojis_from_unicode_org():
    url = "https://unicode.org/Public/emoji/15.1/emoji-test.txt"
    try:
        response = requests.get(url)
        response.raise_for_status()  # Raises an HTTPError if the status is 4xx or 5xx
    except requests.exceptions.RequestException as e:
        print(f"Error fetching emoji data: {e}")
        return {}

    emoji_data = {}
    for line in response.text.splitlines():
        line = line.strip()
        # Skip comments and blank lines
        if not line or line.startswith('#'):
            continue

        # Each line contains the codepoint(s), status, and name
        if '; fully-qualified' in line or '; minimally-qualified' in line:
            parts = line.split(';')
            codepoints = parts[0].strip().replace(' ', '')
            description = parts[-1].split('#')[1].strip()

            # Convert hexadecimal codepoint to character
            # For sequences, this needs to be done point-by-point
            emoji_char = ''.join(chr(int(cp, 16)) for cp in codepoints.split())
            emoji_data[codepoints] = (emoji_char, description)
    return emoji_data

# Example usage
all_emojis = get_emojis_from_unicode_org()
if all_emojis:
    print(f"Found {len(all_emojis)} emojis.")
    # Print the first 10 for demonstration
    for codepoint, (char, desc) in list(all_emojis.items())[:10]:
        print(f"Emoji: {char} | Unicode: {codepoint} | Description: {desc}")
```

```python
class Account:
    """Encapsulated Account class with private attributes"""
    __account_counter = 1001  # Class variable for auto-incrementing account numbers
    
    def __init__(self, name, initial_amount, gender):
        """Initialize account with validation"""
        if initial_amount < 0:
            raise ValueError("Initial amount must be at least $0")
        
        self.__account_number = Account.__account_counter
        Account.__account_counter += 1
        self.__name = name
        self.__balance = initial_amount
        self.__gender = gender
    
    # Getter methods
    def get_account_number(self):
        return self.__account_number
    
    def get_name(self):
        return self.__name
    
    def get_balance(self):
        return self.__balance
    
    def get_gender(self):
        return self.__gender
    
    def deposit(self, amount):
        """Deposit money into account"""
        if amount <= 0:
            print("ERROR: Deposit amount must be positive.")
            return False
        
        self.__balance += amount
        print(f"SUCCESS: Deposited ${amount:,.2f}. New balance: ${self.__balance:,.2f}")
        return True
    
    def withdraw(self, amount):
        """Withdraw money with minimum balance and fee logic"""
        if amount <= 0:
            print("ERROR: Withdrawal amount must be positive.")
            return False
        
        if amount > self.__balance:
            print("ERROR: Insufficient funds.")
            return False
        
        # Check if withdrawal would trigger low balance fee
        balance_after_withdrawal = self.__balance - amount
        
        if balance_after_withdrawal < 500:
            # Check if balance after fee would be negative
            if balance_after_withdrawal - 500 < 0:
                print(f"ERROR: Transaction denied. Withdrawal of ${amount:,.2f} would trigger a $500 fee, resulting in a negative balance.")
                return False
            
            # Apply withdrawal and fee
            self.__balance = balance_after_withdrawal - 500
            print(f"WARNING: Low balance detected. $500 fee applied.")
            print(f"SUCCESS: Withdrew ${amount:,.2f}. New balance: ${self.__balance:,.2f}")
            return True
        
        # Normal withdrawal
        self.__balance -= amount
        print(f"SUCCESS: Withdrew ${amount:,.2f}. New balance: ${self.__balance:,.2f}")
        return True
    
    def transfer(self, recipient_account, amount):
        """Transfer money to another account"""
        if amount <= 0:
            print("ERROR: Transfer amount must be positive.")
            return False
        
        if amount > self.__balance:
            print("ERROR: Insufficient funds for transfer.")
            return False
        
        # Check if transfer would trigger low balance fee
        balance_after_transfer = self.__balance - amount
        
        if balance_after_transfer < 500:
            # Check if balance after fee would be negative
            if balance_after_transfer - 500 < 0:
                print(f"ERROR: Transaction denied. Transfer of ${amount:,.2f} would trigger a $500 fee, resulting in a negative balance.")
                return False
            
            # Apply transfer and fee
            self.__balance = balance_after_transfer - 500
            recipient_account.deposit(amount)
            print(f"WARNING: Low balance detected. $500 fee applied.")
            print(f"SUCCESS: Transferred ${amount:,.2f} to A/C {recipient_account.get_account_number()}.")
            print(f"Your new balance: ${self.__balance:,.2f}")
            return True
        
        # Normal transfer
        self.__balance -= amount
        recipient_account.deposit(amount)
        print(f"SUCCESS: Transferred ${amount:,.2f} to A/C {recipient_account.get_account_number()}.")
        print(f"Your new balance: ${self.__balance:,.2f}")
        return True
    
    def display_info(self):
        """Display account information"""
        print(f"A/C No: {self.__account_number} | Name: {self.__name} | Balance: ${self.__balance:,.2f} | Gender: {self.__gender}")


class Bank:
    """Bank class to manage multiple accounts"""
    def __init__(self):
        self.accounts = {}  # Dictionary to store accounts {account_number: Account object}
    
    def create_account(self, name, initial_amount, gender):
        """Create a new account"""
        try:
            account = Account(name, initial_amount, gender)
            self.accounts[account.get_account_number()] = account
            print("\nACCOUNT CREATED SUCCESSFULLY!")
            account.display_info()
            print("-" * 40)
            return account
        except ValueError as e:
            print(f"ERROR: {e}")
            return None
    
    def find_account(self, account_number):
        """Find account by account number"""
        return self.accounts.get(account_number)
    
    def run_menu(self):
        """Interactive menu-driven interface"""
        while True:
            print("\n" + "=" * 50)
            print("        BANKING SYSTEM MENU")
            print("=" * 50)
            print("1. Create Account")
            print("2. Deposit")
            print("3. Withdraw")
            print("4. Check Balance")
            print("5. Transfer Funds")
            print("6. List All Accounts")
            print("0. Exit")
            print("=" * 50)
            
            try:
                choice = input("Enter your choice: ").strip()
                
                if choice == "1":
                    # Create Account
                    name = input("Enter name: ").strip()
                    gender = input("Enter gender (M/F): ").strip().upper()
                    initial_amount = float(input("Enter initial deposit amount: "))
                    self.create_account(name, initial_amount, gender)
                
                elif choice == "2":
                    # Deposit
                    acc_num = int(input("Enter account number: "))
                    account = self.find_account(acc_num)
                    if account:
                        amount = float(input("Enter deposit amount: "))
                        account.deposit(amount)
                    else:
                        print("ERROR: Account not found.")
                
                elif choice == "3":
                    # Withdraw
                    acc_num = int(input("Enter account number: "))
                    account = self.find_account(acc_num)
                    if account:
                        amount = float(input("Enter withdrawal amount: "))
                        account.withdraw(amount)
                    else:
                        print("ERROR: Account not found.")
                
                elif choice == "4":
                    # Check Balance
                    acc_num = int(input("Enter account number: "))
                    account = self.find_account(acc_num)
                    if account:
                        print(f"\n--- Balance Check (A/C {acc_num}) ---")
                        print(f"A/C {acc_num} Balance: ${account.get_balance():,.2f}")
                    else:
                        print("ERROR: Account not found.")
                
                elif choice == "5":
                    # Transfer Funds
                    from_acc_num = int(input("Enter your account number: "))
                    from_account = self.find_account(from_acc_num)
                    if from_account:
                        to_acc_num = int(input("Enter recipient account number: "))
                        to_account = self.find_account(to_acc_num)
                        if to_account:
                            amount = float(input("Enter transfer amount: "))
                            from_account.transfer(to_account, amount)
                        else:
                            print("ERROR: Recipient account not found.")
                    else:
                        print("ERROR: Your account not found.")
                
                elif choice == "6":
                    # List All Accounts
                    if self.accounts:
                        print("\n--- All Accounts ---")
                        for account in self.accounts.values():
                            account.display_info()
                    else:
                        print("No accounts found.")
                
                elif choice == "0":
                    print("\nThank you for using our banking system!")
                    break
                
                else:
                    print("ERROR: Invalid choice. Please try again.")
            
            except ValueError:
                print("ERROR: Invalid input. Please enter valid data.")
            except Exception as e:
                print(f"ERROR: {e}")


# Sample Test Simulation
if __name__ == "__main__":
    bank = Bank()
    
    # Create accounts
    acc1 = bank.create_account("Amar Panchal", 50000, "M")
    acc2 = bank.create_account("Priya Sharma", 10000, "F")
    
    # Test deposit
    acc1.deposit(1000)
    
    # Test withdrawal
    acc2.withdraw(5000)
    
    # Check balance
    print(f"\n--- Balance Check (A/C 1001) ---")
    print(f"A/C 1001 Balance: ${acc1.get_balance():,.2f}")
    
    # Test fee denial
    print(f"\n--- Fee Trigger Test (A/C 1002 - FAIL) ---")
    acc2.withdraw(4900)
    
    # Test successful fee application
    acc2.deposit(1000)
    acc2.withdraw(5000)
    acc2.withdraw(500)
    acc2.withdraw(1)
    
    # Test transfer
    print(f"\n--- Transfer Test (A/C 1001 -> A/C 1002) ---")
    acc1.transfer(acc2, 10000)
    
    # Final balances
    print(f"\n--- Final Balances ---")
    print(f"A/C 1001 Final Balance: ${acc1.get_balance():,.2f}")
    print(f"A/C 1002 Final Balance: ${acc2.get_balance():,.2f}")
    
    # Uncomment to run interactive menu
    # bank.run_menu()
```

```python
class DS_Example:
    def __init__(self):
        self.stack = []
        # Created an empty stack.

    def push(self, e):
        self.stack.append(e)
        print(f"{e} pushed on stack")

    def pop(self):
        if not self.is_empty():
            popped = self.stack.pop()
            print(f"{popped} popped from stack")
        else:
            print("stack is empty! nothing to pop.")

    def peek(self):
        if not self.is_empty():
            print(f"top element: {self.stack[-1]}")
        else:
            print("stack is empty! no top element.")

    def print_stack(self):  # in LIFO
        if not self.is_empty():
            print("stack elements (top to bottom):")
            for index in range(len(self.stack) - 1, -1, -1):
                print(self.stack[index])
        else:
            print("stack is empty!")

    def is_empty(self):
        return len(self.stack) == 0


# ------------------------------
# main menu driven part
# ------------------------------
ds = DS_Example()

while True:
    print("\n===== STACK MENU =====")
    print("1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Print Stack")
    print("5. Exit")
    
    choice = input("enter your choice (1-5): ")

    if choice == '1':
        element = input("enter element to push: ")
        ds.push(element)
    elif choice == '2':
        ds.pop()
    elif choice == '3':
        ds.peek()
    elif choice == '4':
        ds.print_stack()
    elif choice == '5':
        print("exiting program... 👋")
        break
    else:
        print("invalid choice! please enter between 1–5.")
```

```python
def check_balanced(s):
    stack = []
    for char in s:
        if char == '(':
            stack.append(char)
        elif char == ')':
            if len(stack) == 0:
                return False
            stack.pop()
    if len(stack) == 0:
        return True
    else:
        return False

print(check_balanced("((()))"))
print(check_balanced("(()"))
```

Problem 3: Evaluate postfix expression
Input: "23+" Output: 5
Input: "53*2+" Output: 17

```python
-> eval postfix

def evaluate_postfix(expression):
    stack = []
    for char in expression:
        if char.isdigit():
            stack.append(int(char))
        else:
            b = stack.pop()
            a = stack.pop()
            if char == '+':
                stack.append(a + b)
            elif char == '-':
                stack.append(a - b)
            elif char == '*':
                stack.append(a * b)
            elif char == '/':
                stack.append(a / b)
    return stack.pop()

print(evaluate_postfix("23+"))
print(evaluate_postfix("53*2+"))
```

Problem 4: Check if brackets are valid (multiple types)
Input: "({[]})" Output: True
Input: "({[}])" Output: False

```python
def valid_brackets(s):
    stack = []
    pairs = {'(': ')', '{': '}', '[': ']'}
    for char in s:
        if char in pairs:
            stack.append(char)
        else:
            if len(stack) == 0:
                return False
            top = stack.pop()
            if pairs[top] != char:
                return False
    if len(stack) == 0:
        return True
    else:
        return False

print(valid_brackets("({[]})"))
print(valid_brackets("({[}])"))
```

Problem 5: Interleave two halves of a list
Input: [1,2,3,4,5,6] Output: [1,4,2,5,3,6]

```python
-> interleave

def interleave(arr):
    n = len(arr)
    mid = n // 2
    first_half = []
    second_half = []
    for i in range(mid):
        first_half.append(arr[i])
    for i in range(mid, n):
        second_half.append(arr[i])
    result = []
    i = 0
    while i < mid:
        result.append(first_half[i])
        result.append(second_half[i])
        i = i + 1
    return result

print(interleave([1,2,3,4,5,6]))
```

Problem 3: Stock Span Calculator
You are analyzing stock prices. For each day, calculate the "span" - the maximum number of consecutive days before today where the price was less than or equal to today's price.
Input: prices = [100, 80, 60, 70, 60, 75, 85]
Output: spans = [1, 1, 1, 2, 1, 4, 6]
Explanation: Day 6 (price=85) has span 6 because all previous days had price <= 85

```python
-> stock span

def calculate_span(prices):
    n = len(prices)
    spans = []
    stack = []
    
    for i in range(n):
        span = 1
        while len(stack) > 0 and prices[stack[-1]] <= prices[i]:
            stack.pop()
        if len(stack) == 0:
            span = i + 1
        else:
            span = i - stack[-1]
        spans.append(span)
        stack.append(i)
    
    return spans

prices = [100, 80, 60, 70, 60, 75, 85]
print(calculate_span(prices))
```

Problem 2: Sliding Window Maximum in a Stream
You are monitoring a data stream and need to track the maximum value in the last k readings at any point. As new data arrives, old data outside the window is discarded.
Input: stream = [1,3,-1,-3,5,3,6,7], k=3
Output: maximums = [3, 3, 5, 5, 6, 7]
Explanation: Window [1,3,-1] max=3, [3,-1,-3] max=3, etc.

```python
-> slid win

def sliding_window_max(stream, k):
    result = []
    queue = []
    
    for i in range(len(stream)):
        while len(queue) > 0 and stream[queue[-1]] < stream[i]:
            queue.pop()
        queue.append(i)
        if queue[0] < i - k + 1:
            queue.pop(0)
        if i >= k - 1:
            result.append(stream[queue[0]])
    
    return result

stream = [1,3,-1,-3,5,3,6,7]
k = 3
print(sliding_window_max(stream, k))
```

```python
-> circ gas

def gas_station(gas, cost):
    n = len(gas)
    total_gas = 0
    total_cost = 0
    tank = 0
    start = 0
    
    for i in range(n):
        total_gas = total_gas + gas[i]
        total_cost = total_cost + cost[i]
        tank = tank + gas[i] - cost[i]
        if tank < 0:
            start = i + 1
            tank = 0
    
    if total_gas < total_cost:
        return -1
    else:
        return start

gas = [1,2,3,4,5]
cost = [3,4,5,1,2]
print(gas_station(gas, cost))
```

Problem 1: Clone a Complex Network with Random Pointers
You have a network where each node has a value, a next pointer, and a random pointer that can point to any node in the network or None. Create a deep copy of this entire network structure preserving all connections.

Output: A completely independent copy with same structure

```python
-> clone net

class RandomNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.random = None

def clone_network(head):
    if head == None:
        return None
    
    current = head
    while current != None:
        new_node = RandomNode(current.data)
        new_node.next = current.next
        current.next = new_node
        current = new_node.next
    
    current = head
    while current != None:
        if current.random != None:
            current.next.random = current.random.next
        current = current.next.next
    
    original = head
    copy = head.next
    copy_head = copy
    while original != None:
        original.next = original.next.next
        if copy.next != None:
            copy.next = copy.next.next
        original = original.next
        copy = copy.next
    
    return copy_head

head = RandomNode(1)
head.next = RandomNode(2)
head.next.next = RandomNode(3)
head.random = head.next.next
head.next.random = head
cloned = clone_network(head)
print("Original:", head.data, "Cloned:", cloned.data)
```

Problem 2: LRU Cache Implementation
Design a data structure for Least Recently Used (LRU) cache that supports get and put operations in O(1) time. When cache reaches capacity, it should invalidate the least recently used item before inserting a new item.
Input: capacity=2, operations: put(1,1), put(2,2), get(1), put(3,3), get(2)
Output: get(1)=1, get(2)=-1 (evicted)

```python
class LRUNode:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}
        self.head = LRUNode(0, 0)
        self.tail = LRUNode(0, 0)
        self.head.next = self.tail
        self.tail.prev = self.head
    
    def remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev
    
    def add_to_head(self, node):
        node.next = self.head.next
        node.prev = self.head
        self.head.next.prev = node
        self.head.next = node
    
    def get(self, key):
        if key in self.cache:
            node = self.cache[key]
            self.remove(node)
            self.add_to_head(node)
            return node.value
        return -1
    
    def put(self, key, value):
        if key in self.cache:
            self.remove(self.cache[key])
        node = LRUNode(key, value)
        self.add_to_head(node)
        self.cache[key] = node
        if len(self.cache) > self.capacity:
            lru = self.tail.prev
            self.remove(lru)
            del self.cache[lru.key]

cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print(cache.get(1))
cache.put(3, 3)
print(cache.get(2))
```

Problem 3: Flatten a Multilevel Doubly Linked List
You have a special doubly linked list where each node may have a child pointer to another doubly linked list. Flatten this structure into a single-level doubly linked list where child lists are inserted in place.
Input: 1<->2<->3<->4 where 2 has child 5<->6 and 5 has child 7<->8
Output: 1<->2<->5<->7<->8<->6<->3<->4

```python
class MultiNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None
        self.child = None

def flatten(head):
    if head == None:
        return head
    
    current = head
    while current != None:
        if current.child != None:
            child_tail = current.child
            while child_tail.next != None:
                child_tail = child_tail.next
            child_tail.next = current.next
            if current.next != None:
                current.next.prev = child_tail
            current.next = current.child
            current.child.prev = current
            current.child = None
        current = current.next
    
    return head

def print_flat(head):
    current = head
    while current != None:
        print(current.data, end="<->")
        current = current.next
    print("None")

head = MultiNode(1)
head.next = MultiNode(2)
head.next.prev = head
head.next.next = MultiNode(3)
head.next.next.prev = head.next
child = MultiNode(5)
child.next = MultiNode(6)
child.next.prev = child
head.next.child = child
flattened = flatten(head)
print_flat(flattened)
```
