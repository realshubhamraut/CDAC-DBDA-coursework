```python
dd,mm,yyyy = map(int, input("Enter 3 nums").split('/'))
if mm == 1:
    month_name = 'jan'
elif mm == 2:
    month_name = 'feb'
elif mm == 3:
    month_name = 'march'
elif mm == 4:
    month_name = 'apr'
elif mm == 5:
    month_name = 'may'
else:
    print('Invalid month')
print(f"Date: {dd}/{month_name}/{yyyy}")
```

```python
# findin if nums are imaginary roots
a,b,c = map(float, input('enter 3 variables').split(","))
if a and b and c == 0:
    print('not a quadratic eqn')

d = (b**2) - (4*a*c)
if d <=0:
    print('imaginary roots')

else:
    root1 = (-b+(d)**0.5)/ (2*a)
    root2 = (-b-(d)**0.5)/ (2*a)
    print(f"root1:{root1}, root2:{root2}")
```
