```python
# #1 


# you are given a list of integers where every elements appears an even number f times except one 

# write a progream to find the element that appears an odd number of times

# example

# input [1,2,3,2,3,1,3] -> output 3
    
```

```python
def find_odd_occurrence(nums):
    result = 0
    for n in nums:
        result ^= n
    return result

nums = [1, 2, 3, 2, 3, 1, 3]
print(find_odd_occurrence(nums))
```

```python
# given a  list of integers and a target number, find a continuous subarray that adds up exactly to the target

# return the subarray or indicate if none exists

# example input -> [4,2,7,1,9,5]  target = 10 output -> [2,7,1]
```

```python
def find_subarray_sum(nums, target):
    start = 0
    curr_sum = 0

    for end in range(len(nums)):
        curr_sum += nums[end]

        # shrink the window until the sum <= target
        while curr_sum > target and start <= end:
            curr_sum -= nums[start]
            start += 1

        if curr_sum == target:
            return nums[start:end + 1]

    return None

nums = [4, 2, 7, 1, 9, 5]
target = 10
print(find_subarray_sum(nums, target))
```

```python
write a function that moves all zeros in a list ot the end while maintaining the order of non-zero elements

input -> [0,1,0,3,12] -> output -> [1,2,12,0,0]
```

```python
def move_zeros(nums):
    pos = 0  # position for next non-zero element

    for i in range(len(nums)):
        if nums[i] != 0:
            nums[pos], nums[i] = nums[i], nums[pos]
            pos += 1
    return nums

# example
nums = [0, 1, 0, 3, 12]
print(move_zeros(nums))   # output: [1, 3, 12, 0, 0]
```

```python
def move_zeros(nums):
    result = [x for x in nums if x != 0]  # keep non-zero
    result += [0] * (len(nums) - len(result))  # fill zeros at end
    return result

print(move_zeros([0, 1, 0, 3, 12]))  # [1, 3, 12, 0, 0]
```
