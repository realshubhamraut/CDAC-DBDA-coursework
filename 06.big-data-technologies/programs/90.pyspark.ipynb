```python
# textFile → split → map → reduceByKey → sort → save
```

```python
from pyspark import SparkContext
```

```python
sc = SparkContext("local", "WordCount Example")
```

```python
rdd = sc.textFile("sample.txt")  # Replace with your file path
```

```python
words = rdd.flatMap(lambda line: line.split())
```

```python
word_pairs = words.map(lambda word: (word, 1))
```

```python
word_counts = word_pairs.reduceByKey(lambda a, b: a + b)
```

```python
sorted_counts = word_counts.sortBy(lambda x: x[1], ascending=False)
```

```python
sorted_counts.saveAsTextFile("output_wordcount")
```

```python
sc.stop()
```
