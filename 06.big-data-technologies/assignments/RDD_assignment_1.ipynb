# question - 3: Day of the week wise sales (for e.g. all Sunday together). Print Sunday Onwards..


```python
rdd1 = sc.textFile("/content/sample_data/txn.txt")
```

```python
from datetime import datetime
```

```python
format_date = "%m-%d-%Y"
```

```python
rdd2 = rdd1.map(lambda a : (datetime.strptime(a.split(',')[1], format_date).strftime("%w") + " " + (datetime.strptime(a.split(',')[1], format_date).strftime("%A")), float(a.split(",")[3])))
```

```python
for var in rdd2.take(5):
  print(var)
```

```python
rdd3 = rdd2.reduceByKey(lambda a,b : round(a + b, 2))
```

```python
rdd3.count()
```

```python
for var in rdd3.collect():
  print(var)
```

```python
rdd4 = rdd3.sortByKey()
```

```python
for var in rdd4.collect():
  print(var)
```

```python
csv_output = rdd4.map(lambda a : a[0].split(' ')[1] + ',' + str(round(a[1],2)))
```

```python
for var in csv_output.collect():
  print(var)
```

```python
csv = csv_output.coalesce(1)
```

```python
csv.getNumPartitions()
```

```python
csv.saveAsTextFile('/content/sample_data/Data/Day_wise_sales')
```
